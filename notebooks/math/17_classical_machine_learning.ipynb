## 1. Learning setup

Empirical risk minimization: $\hat f = \arg\min_f\frac1n\sum_i\ell(f(x_i), y_i) + \lambda\Omega(f)$

Train/validation/test split; $k$-fold cross-validation estimate $\frac1k\sum_j\text{err}_j$.


## 2. Linear regression

Model $\hat y = Xw + b$, loss $\frac1n\|y - Xw\|^2$.

**Normal equation:** $w = (X^\top X)^{-1}X^\top y$ (use SVD / `lstsq` in practice). Gradient $\nabla_w = -\frac2nX^\top(y - Xw)$.

**Ridge (L2):** $\min\|y - Xw\|^2 + \lambda\|w\|^2\Rightarrow w = (X^\top X + \lambda I)^{-1}X^\top y$

**Lasso (L1):** $\min\frac1{2n}\|y - Xw\|^2 + \lambda\|w\|_1$; coordinate update $w_j\leftarrow\frac{S(\rho_j,\lambda)}{z_j}$ with soft-thresholding $S(\rho,\lambda) = \text{sign}(\rho)\max(|\rho| - \lambda, 0)$

**Elastic net:** $\lambda_1\|w\|_1 + \lambda_2\|w\|_2^2$

**Bayesian linear regression:** prior $w\sim\mathcal N(0,\alpha^{-1}I)$, noise precision $\beta$: posterior $\Sigma_N = (\alpha I + \beta X^\top X)^{-1}$, $\mu_N = \beta\Sigma_NX^\top y$; predictive variance $\frac1\beta + x^\top\Sigma_Nx$

**Polynomial / basis-function regression:** replace $x$ by $\phi(x) = (1, x, x^2, \dots)$.

**Huber loss:** $\ell_\delta(r) = \frac12r^2$ if $|r|\le\delta$, else $\delta(|r| - \frac12\delta)$. **Quantile (pinball) loss:** $\ell_\tau(r) = \max(\tau r, (\tau - 1)r)$.


## 3. Logistic and softmax regression

$P(y = 1|x) = \sigma(w^\top x + b)$; log-odds $\log\frac{p}{1-p} = w^\top x + b$

Binary cross-entropy $L = -\frac1n\sum[y\log p + (1-y)\log(1-p)]$; gradient $\nabla_w = \frac1nX^\top(p - y)$; Hessian $X^\top\text{diag}(p(1-p))X$ (Newton's method here is called IRLS)

Softmax regression: $P(y = k|x) = \frac{e^{w_k^\top x}}{\sum_je^{w_j^\top x}}$; gradient $\nabla_{W} = \frac1nX^\top(P - Y)$

**Generalized linear models:** $\mathbb E[y|x] = g^{-1}(w^\top x)$ with link $g$: identity (Gaussian), logit (Bernoulli), log (Poisson).


## 4. Perceptron

Predict $\hat y = \text{sign}(w^\top x)$; on a mistake $w\leftarrow w + y_ix_i$. **Novikoff's theorem:** with margin $\gamma$ and $\|x\|\le R$, at most $(R/\gamma)^2$ mistakes.


## 5. Support vector machines

**Hard margin:** $\min\frac12\|w\|^2$ s.t. $y_i(w^\top x_i + b)\ge1$; margin $= \frac{2}{\|w\|}$

**Soft margin:** $\min\frac12\|w\|^2 + C\sum\xi_i$ s.t. $y_i(w^\top x_i + b)\ge1 - \xi_i$, $\xi_i\ge0$; equivalently hinge loss $\sum\max(0, 1 - y_if(x_i)) + \frac{\lambda}{2}\|w\|^2$

**Dual:** $\max_\alpha\sum\alpha_i - \frac12\sum_{ij}\alpha_i\alpha_jy_iy_jx_i^\top x_j$ s.t. $0\le\alpha_i\le C$, $\sum\alpha_iy_i = 0$; then $w = \sum\alpha_iy_ix_i$, and points with $\alpha_i > 0$ are the support vectors.

**Kernel trick:** replace $x_i^\top x_j$ with $k(x_i, x_j)$; decision $f(x) = \sum\alpha_iy_ik(x_i, x) + b$.

Kernels: linear $x^\top z$; polynomial $(\gamma x^\top z + r)^d$; RBF $\exp(-\gamma\|x - z\|^2)$; sigmoid $\tanh(\gamma x^\top z + r)$

**Representer theorem:** the minimizer of a regularized loss in an RKHS has the form $f = \sum\alpha_ik(x_i,\cdot)$.

**SVR:** $\varepsilon$-insensitive loss $\max(0, |y - f(x)| - \varepsilon)$.


## 6. Kernel and Gaussian-process methods

**Kernel ridge regression:** $\alpha = (K + \lambda I)^{-1}y$, $f(x) = k(x)^\top\alpha$

**Gaussian process regression:** with $K = k(X,X)$, $k_* = k(X, x_*)$:
$$\mu_* = k_*^\top(K + \sigma_n^2I)^{-1}y, \qquad \sigma_*^2 = k(x_*,x_*) - k_*^\top(K + \sigma_n^2I)^{-1}k_*$$
Log marginal likelihood $-\frac12y^\top(K+\sigma^2I)^{-1}y - \frac12\log|K + \sigma^2I| - \frac n2\log2\pi$ (to fit kernel hyperparameters).

**Random Fourier features:** $k(x,z)\approx\phi(x)^\top\phi(z)$, $\phi(x) = \sqrt{\frac2D}\cos(\Omega^\top x + b)$, $\Omega\sim\mathcal N(0, \sigma^{-2}I)$ for the RBF kernel.


## 7. Probabilistic classifiers

**Naive Bayes:** $P(y|x)\propto P(y)\prod_jP(x_j|y)$. Gaussian: $P(x_j|y) = \mathcal N(\mu_{jy}, \sigma_{jy}^2)$. Multinomial with Laplace smoothing: $P(w|y) = \frac{\text{count}(w,y) + \alpha}{\sum_{w'}\text{count}(w', y) + \alpha|V|}$

**LDA (shared covariance $\Sigma$):** $\delta_k(x) = x^\top\Sigma^{-1}\mu_k - \frac12\mu_k^\top\Sigma^{-1}\mu_k + \log\pi_k$ (linear boundary)

**QDA:** $\delta_k(x) = -\frac12\log|\Sigma_k| - \frac12(x - \mu_k)^\top\Sigma_k^{-1}(x - \mu_k) + \log\pi_k$

**Fisher's linear discriminant:** maximize $J(w) = \frac{w^\top S_Bw}{w^\top S_Ww}\Rightarrow w\propto S_W^{-1}(\mu_1 - \mu_2)$


## 8. Nearest neighbours and distances

Euclidean $\|x - z\|_2$; Manhattan $\|x - z\|_1$; Minkowski $\|x - z\|_p$; cosine $1 - \frac{x^\top z}{\|x\|\|z\|}$; Mahalanobis $\sqrt{(x - z)^\top\Sigma^{-1}(x - z)}$; Hamming; Jaccard $1 - \frac{|A\cap B|}{|A\cup B|}$

$k$-NN classification: majority vote of the $k$ nearest; regression: mean (or distance-weighted mean $\sum w_iy_i/\sum w_i$, $w_i = 1/d_i$).

Curse of dimensionality: in high dimensions, distances concentrate ($\frac{d_{\max} - d_{\min}}{d_{\min}}\to0$).


## 9. Decision trees

**Entropy:** $H(S) = -\sum_kp_k\log_2p_k$. **Information gain:** $IG(S, A) = H(S) - \sum_v\frac{|S_v|}{|S|}H(S_v)$ (ID3)

**Gain ratio (C4.5):** $\frac{IG}{\text{SplitInfo}}$, $\text{SplitInfo} = -\sum_v\frac{|S_v|}{|S|}\log_2\frac{|S_v|}{|S|}$

**Gini impurity (CART):** $G = 1 - \sum_kp_k^2$. Regression trees split to minimize the sum of squared errors.

**Cost-complexity pruning:** $R_\alpha(T) = R(T) + \alpha|T|$


## 10. Ensembles

**Bagging:** average $B$ models trained on bootstrap samples; variance of an average of correlated models $\rho\sigma^2 + \frac{1-\rho}{B}\sigma^2$. A bootstrap sample leaves out about $1 - (1 - 1/n)^n\approx36.8\%$ of points (out-of-bag estimate).

**Random forest:** bagging + random subset of $m\approx\sqrt p$ features per split.

**AdaBoost** (labels $\pm1$): weighted error $\varepsilon_t = \sum_iw_i[h_t(x_i)\ne y_i]$, $\;\alpha_t = \frac12\ln\frac{1-\varepsilon_t}{\varepsilon_t}$, $\;w_i\leftarrow w_ie^{-\alpha_ty_ih_t(x_i)}$ (normalized); $H(x) = \text{sign}\sum_t\alpha_th_t(x)$. It minimizes exponential loss $e^{-yf(x)}$.

**Gradient boosting:** $F_m = F_{m-1} + \nu h_m$, where $h_m$ fits the pseudo-residuals $r_i = -\frac{\partial\ell(y_i, F)}{\partial F}\big|_{F_{m-1}}$.

**XGBoost:** second-order Taylor expansion with $g_i, h_i$; optimal leaf weight $w_j^* = -\frac{G_j}{H_j + \lambda}$; split gain $\frac12\left[\frac{G_L^2}{H_L + \lambda} + \frac{G_R^2}{H_R + \lambda} - \frac{(G_L + G_R)^2}{H_L + H_R + \lambda}\right] - \gamma$

**Stacking:** a meta-model trained on out-of-fold predictions of base models.


## 11. Clustering

**k-means:** minimize $\sum_k\sum_{x\in C_k}\|x - \mu_k\|^2$; alternate assignment $c_i = \arg\min_k\|x_i - \mu_k\|^2$ and update $\mu_k = \frac{1}{|C_k|}\sum_{x\in C_k}x$. **k-means++** seeds with probability $\propto D(x)^2$.

**Gaussian mixture model + EM:**
- E-step: $\gamma_{ik} = \frac{\pi_k\mathcal N(x_i|\mu_k,\Sigma_k)}{\sum_j\pi_j\mathcal N(x_i|\mu_j, \Sigma_j)}$
- M-step: $N_k = \sum_i\gamma_{ik}$, $\;\mu_k = \frac1{N_k}\sum_i\gamma_{ik}x_i$, $\;\Sigma_k = \frac1{N_k}\sum_i\gamma_{ik}(x_i - \mu_k)(x_i - \mu_k)^\top$, $\;\pi_k = \frac{N_k}{n}$

**EM in general:** maximize the lower bound $\log p(x|\theta)\ge\mathbb E_{q(z)}[\log p(x,z|\theta)] - \mathbb E_q[\log q(z)]$, with $q(z) = p(z|x,\theta^{old})$ in the E-step.

**Hierarchical linkage:** single $\min d(a,b)$, complete $\max d(a,b)$, average, Ward (minimum increase in within-cluster variance).

**DBSCAN:** core point has $\ge$ MinPts neighbours within $\varepsilon$; clusters are density-connected core points.

**Spectral clustering:** take the $k$ smallest eigenvectors of the normalized Laplacian, then run k-means on the rows.

**Evaluation:** silhouette $s = \frac{b - a}{\max(a,b)}$; Davies-Bouldin; Calinski-Harabasz; adjusted Rand index; normalized mutual information.


## 12. Dimensionality reduction

**PCA:** centred data $X$, covariance $S = \frac{1}{n-1}X^\top X = V\Lambda V^\top$; project $Z = XV_k$. Via SVD $X = U\Sigma V^\top$, $\lambda_i = \frac{\sigma_i^2}{n-1}$. Explained variance ratio $\frac{\lambda_i}{\sum_j\lambda_j}$. PCA maximizes projected variance and minimizes reconstruction error.

**Kernel PCA:** eigendecompose the centred Gram matrix $\tilde K = K - \mathbf 1_nK - K\mathbf 1_n + \mathbf 1_nK\mathbf 1_n$.

**ICA:** find $W$ so that $s = Wx$ has independent (maximally non-Gaussian) components; FastICA update $w\leftarrow\mathbb E[xg(w^\top x)] - \mathbb E[g'(w^\top x)]w$.

**NMF:** $X\approx WH$, $W,H\ge0$; multiplicative updates $H\leftarrow H\odot\frac{W^\top X}{W^\top WH}$, $W\leftarrow W\odot\frac{XH^\top}{WHH^\top}$

**Classical MDS:** double-centre squared distances $B = -\frac12JD^{(2)}J$, embed with top eigenvectors.

**t-SNE:** $p_{j|i}\propto\exp(-\|x_i - x_j\|^2/2\sigma_i^2)$, $p_{ij} = \frac{p_{j|i} + p_{i|j}}{2n}$; $q_{ij}\propto(1 + \|y_i - y_j\|^2)^{-1}$; minimize $KL(P\|Q)$; gradient $4\sum_j(p_{ij} - q_{ij})(y_i - y_j)(1 + \|y_i - y_j\|^2)^{-1}$

**UMAP:** fuzzy simplicial-set graph with cross-entropy between high- and low-dimensional edge weights.

**Johnson-Lindenstrauss lemma:** $n$ points can be projected to $k = O(\varepsilon^{-2}\log n)$ dimensions while preserving pairwise distances within $1\pm\varepsilon$.


## 13. Hidden Markov models

Parameters: transitions $A_{ij}$, emissions $B_j(o)$, initial $\pi$.

**Forward:** $\alpha_t(j) = \left[\sum_i\alpha_{t-1}(i)A_{ij}\right]B_j(o_t)$, $\;P(O) = \sum_j\alpha_T(j)$

**Backward:** $\beta_t(i) = \sum_jA_{ij}B_j(o_{t+1})\beta_{t+1}(j)$

**Viterbi:** $\delta_t(j) = \max_i\delta_{t-1}(i)A_{ij}B_j(o_t)$, with back-pointers

**Baum-Welch (EM):** $\gamma_t(i) = \frac{\alpha_t(i)\beta_t(i)}{P(O)}$, $\;\xi_t(i,j) = \frac{\alpha_t(i)A_{ij}B_j(o_{t+1})\beta_{t+1}(j)}{P(O)}$, $\;A_{ij}\leftarrow\frac{\sum_t\xi_t(i,j)}{\sum_t\gamma_t(i)}$


## 14. Graphical models and approximate inference

**Bayesian network:** $p(x) = \prod_ip(x_i|\text{pa}(x_i))$. **Markov random field:** $p(x) = \frac1Z\prod_C\psi_C(x_C)$.

**Belief propagation (sum-product):** $m_{i\to j}(x_j) = \sum_{x_i}\psi_{ij}(x_i, x_j)\psi_i(x_i)\prod_{k\in N(i)\setminus j}m_{k\to i}(x_i)$

**Linear-chain CRF:** $p(y|x) = \frac1{Z(x)}\exp\left(\sum_t\sum_k\lambda_kf_k(y_{t-1}, y_t, x, t)\right)$

**Variational inference:** maximize ELBO $\mathcal L(q) = \mathbb E_q[\log p(x,z)] - \mathbb E_q[\log q(z)] = \log p(x) - KL(q\|p(z|x))$; mean-field CAVI update $\log q_j^*(z_j) = \mathbb E_{-j}[\log p(x,z)] + \text{const}$

**LDA topic model:** $\theta_d\sim\text{Dir}(\alpha)$, $z_{dn}\sim\text{Cat}(\theta_d)$, $w_{dn}\sim\text{Cat}(\beta_{z_{dn}})$; collapsed Gibbs $P(z = k|\cdot)\propto(n_{dk} + \alpha)\frac{n_{kw} + \beta}{n_k + V\beta}$


## 15. Recommender systems and association rules

**Matrix factorization:** $\min\sum_{(u,i)\in\Omega}(r_{ui} - p_u^\top q_i - b_u - b_i - \mu)^2 + \lambda(\|p_u\|^2 + \|q_i\|^2)$; ALS alternates closed-form ridge solutions.

**Factorization machine:** $\hat y = w_0 + \sum w_ix_i + \sum_{i<j}\langle v_i, v_j\rangle x_ix_j$

**Association rules:** support $= P(A\cup B)$, confidence $= P(B|A)$, lift $= \frac{P(B|A)}{P(B)}$


## 16. Evaluation metrics

Confusion matrix: TP, FP, TN, FN.

Accuracy $\frac{TP + TN}{N}$; precision $\frac{TP}{TP + FP}$; recall (sensitivity, TPR) $\frac{TP}{TP + FN}$; specificity $\frac{TN}{TN + FP}$; FPR $= 1 -$ specificity

$F_1 = \frac{2PR}{P + R}$; $F_\beta = (1 + \beta^2)\frac{PR}{\beta^2P + R}$; Matthews correlation $\frac{TP\cdot TN - FP\cdot FN}{\sqrt{(TP+FP)(TP+FN)(TN+FP)(TN+FN)}}$

ROC-AUC = probability that a random positive scores higher than a random negative. PR-AUC / average precision for imbalanced data.

Log loss; Brier score $\frac1n\sum(p_i - y_i)^2$; expected calibration error $\sum_b\frac{|B_b|}{n}|\text{acc}(B_b) - \text{conf}(B_b)|$

Ranking: precision@k, recall@k, MRR $= \frac1{|Q|}\sum\frac{1}{\text{rank}_i}$, DCG $= \sum_i\frac{2^{rel_i} - 1}{\log_2(i + 1)}$, NDCG $= \frac{DCG}{IDCG}$

Regression: MAE, MSE, RMSE, $R^2$, MAPE.


## 17. Statistical learning theory (short form; more in 25)

Bias-variance trade-off (see 12 §8). **PAC bound (finite hypothesis class):** with probability $1 - \delta$, $R(h)\le\hat R(h) + \sqrt{\frac{\ln|\mathcal H| + \ln(1/\delta)}{2n}}$.
