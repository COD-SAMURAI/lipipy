# 09 · Integral Transforms: Laplace, Fourier and Z

- **Level:** Intermediate
- **Typical semester:** Semester 3 (Engineering Mathematics III) and signals and systems courses
- **Prerequisites:** 03, 04, 05, 08
- **Used later in:** convolutions (CNNs), FFT-based long convolutions (Hyena, FNO), state-space model discretization, spectral graph theory.


## 1. Laplace transform: definition and standard transforms

$$\mathcal L\{f(t)\} = F(s) = \int_0^\infty e^{-st}f(t)\,dt$$

| $f(t)$ | $F(s)$ | $f(t)$ | $F(s)$ |
|---|---|---|---|
| $1$ | $\frac1s$ | $\sin at$ | $\frac{a}{s^2+a^2}$ |
| $t^n$ | $\frac{n!}{s^{n+1}}$ | $\cos at$ | $\frac{s}{s^2+a^2}$ |
| $e^{at}$ | $\frac1{s-a}$ | $\sinh at$ | $\frac{a}{s^2-a^2}$ |
| $t^ne^{at}$ | $\frac{n!}{(s-a)^{n+1}}$ | $\cosh at$ | $\frac{s}{s^2-a^2}$ |
| $u(t-a)$ | $\frac{e^{-as}}{s}$ | $\delta(t-a)$ | $e^{-as}$ |


## 2. Laplace transform properties

**Linearity:** $\mathcal L\{af + bg\} = aF + bG$
**First shifting:** $\mathcal L\{e^{at}f(t)\} = F(s-a)$
**Second shifting:** $\mathcal L\{f(t-a)u(t-a)\} = e^{-as}F(s)$
**Change of scale:** $\mathcal L\{f(at)\} = \frac1aF\left(\frac sa\right)$
**Derivatives:** $\mathcal L\{f'\} = sF - f(0)$, $\;\mathcal L\{f''\} = s^2F - sf(0) - f'(0)$
**Integral:** $\mathcal L\left\{\int_0^tf(\tau)d\tau\right\} = \frac{F(s)}{s}$
**Multiplication by $t$:** $\mathcal L\{t^nf(t)\} = (-1)^n\frac{d^n}{ds^n}F(s)$
**Division by $t$:** $\mathcal L\left\{\frac{f(t)}{t}\right\} = \int_s^\infty F(u)\,du$
**Periodic function** (period $T$): $\mathcal L\{f\} = \frac{1}{1-e^{-sT}}\int_0^Te^{-st}f(t)\,dt$
**Initial/final value:** $f(0^+) = \lim_{s\to\infty}sF(s)$, $\;f(\infty) = \lim_{s\to0}sF(s)$
**Convolution:** $\mathcal L\{(f*g)(t)\} = F(s)G(s)$, where $(f*g)(t) = \int_0^tf(\tau)g(t-\tau)\,d\tau$


## 3. Inverse Laplace and solving ODEs

Inverse transform by partial fractions, shifting theorems, convolution theorem, or the Bromwich integral $f(t) = \frac{1}{2\pi i}\int_{\gamma-i\infty}^{\gamma+i\infty}e^{st}F(s)\,ds$.

To solve a linear ODE: transform both sides, solve algebraically for $Y(s)$, invert. Transfer function $H(s) = \frac{Y(s)}{X(s)}$.


## 4. Fourier transform

$$\hat f(\omega) = \mathcal F\{f\} = \int_{-\infty}^\infty f(x)e^{-i\omega x}dx, \qquad f(x) = \frac{1}{2\pi}\int_{-\infty}^\infty\hat f(\omega)e^{i\omega x}d\omega$$

Fourier sine / cosine transforms: $F_s(\omega) = \int_0^\infty f(x)\sin\omega x\,dx$, $\;F_c(\omega) = \int_0^\infty f(x)\cos\omega x\,dx$

**Properties:** shift $\mathcal F\{f(x-a)\} = e^{-i\omega a}\hat f$; modulation $\mathcal F\{e^{iax}f\} = \hat f(\omega - a)$; scaling $\mathcal F\{f(ax)\} = \frac{1}{|a|}\hat f(\omega/a)$; derivative $\mathcal F\{f'\} = i\omega\hat f$

**Convolution theorem:** $\mathcal F\{f*g\} = \hat f\cdot\hat g$

**Parseval / Plancherel:** $\int|f|^2dx = \frac1{2\pi}\int|\hat f|^2d\omega$

**Gaussian is its own transform:** $\mathcal F\{e^{-x^2/2}\} = \sqrt{2\pi}e^{-\omega^2/2}$

**Uncertainty principle:** $\sigma_x\sigma_\omega\ge\frac12$


## 5. Discrete Fourier transform and FFT

$$X_k = \sum_{n=0}^{N-1}x_ne^{-2\pi ikn/N}, \qquad x_n = \frac1N\sum_{k=0}^{N-1}X_ke^{2\pi ikn/N}$$

The FFT (Cooley-Tukey) computes this in $O(N\log N)$ instead of $O(N^2)$ by splitting into even and odd indices: $X_k = E_k + e^{-2\pi ik/N}O_k$.

Circular convolution in time = pointwise product of DFTs. Linear convolution of length-$N$ and length-$M$ signals: zero-pad to $N+M-1$.

**Sampling theorem (Nyquist-Shannon):** a signal band-limited to $f_{\max}$ is fully recovered from samples at rate $f_s > 2f_{\max}$; otherwise aliasing.

**ML note:** FFT convolutions make long-range convolution models (S4 kernels, Hyena, Fourier neural operators) run in $O(N\log N)$.


## 6. Z-transform

$$Z\{x_n\} = X(z) = \sum_{n=0}^{\infty}x_nz^{-n}$$

| $x_n$ | $X(z)$ |
|---|---|
| $\delta_n$ | $1$ |
| $1$ | $\frac{z}{z-1}$ |
| $a^n$ | $\frac{z}{z-a}$ |
| $n$ | $\frac{z}{(z-1)^2}$ |
| $na^n$ | $\frac{az}{(z-a)^2}$ |
| $\cos n\theta$ | $\frac{z(z-\cos\theta)}{z^2-2z\cos\theta+1}$ |
| $\sin n\theta$ | $\frac{z\sin\theta}{z^2-2z\cos\theta+1}$ |

**Properties:** shift $Z\{x_{n+1}\} = zX(z) - zx_0$, $\;Z\{x_{n-k}\} = z^{-k}X(z)$; damping $Z\{a^nx_n\} = X(z/a)$; convolution $Z\{x*y\} = X(z)Y(z)$

**Initial/final value:** $x_0 = \lim_{z\to\infty}X(z)$, $\;\lim_{n\to\infty}x_n = \lim_{z\to1}(z-1)X(z)$

Used to solve **difference equations** (e.g. $y_{n+2} - 3y_{n+1} + 2y_n = 0$) the way Laplace solves ODEs. A discrete system is stable iff all poles lie inside the unit circle $|z|<1$.

**ML note:** a linear recurrence $h_t = \bar Ah_{t-1} + \bar Bx_t$ (RNNs without nonlinearity, SSMs) is stable iff the eigenvalues of $\bar A$ are inside the unit circle.


## 7. Other transforms worth knowing

**Hilbert transform** (analytic signal), **wavelet transform** $W(a,b) = \frac{1}{\sqrt{|a|}}\int f(t)\psi^*\left(\frac{t-b}{a}\right)dt$ (multi-resolution analysis), **Radon transform** (CT imaging), **Mellin transform**, **discrete cosine transform** (JPEG), **graph Fourier transform** $\hat f = U^\top f$ where $L = U\Lambda U^\top$ is the graph Laplacian (see 23 · Graph learning).
