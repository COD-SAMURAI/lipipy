# 13 · Discrete Mathematics and Graph Theory

- **Level:** Fundamentals → Intermediate
- **Typical semester:** Semester 3–4 (Discrete Mathematical Structures, for CS/IS branches)
- **Prerequisites:** 01
- **Used later in:** algorithm analysis, logic and classical AI, graph neural networks, knowledge graphs, combinatorics in probability.


## 1. Sets, relations and functions

$|A\cup B| = |A| + |B| - |A\cap B|$; De Morgan: $\overline{A\cup B} = \bar A\cap\bar B$, $\;\overline{A\cap B} = \bar A\cup\bar B$; power set $|\mathcal P(A)| = 2^{|A|}$; $|A\times B| = |A||B|$

Relation properties: reflexive, symmetric, antisymmetric, transitive. Equivalence relation = reflexive + symmetric + transitive (gives a partition). Partial order = reflexive + antisymmetric + transitive (Hasse diagrams, lattices).

Functions: injective, surjective, bijective. Number of functions $A\to B$ is $|B|^{|A|}$; onto functions from $m$ to $n$ elements: $\sum_{k=0}^n(-1)^k\binom nk(n-k)^m$.


## 2. Propositional and predicate logic

Connectives $\neg,\wedge,\vee,\to,\leftrightarrow$; $p\to q\equiv\neg p\vee q$; contrapositive $\neg q\to\neg p$

De Morgan: $\neg(p\wedge q)\equiv\neg p\vee\neg q$; distributive, absorption, idempotent laws

Rules of inference: modus ponens ($p,\;p\to q\vdash q$), modus tollens ($\neg q,\;p\to q\vdash\neg p$), hypothetical syllogism, disjunctive syllogism, resolution ($p\vee q,\;\neg p\vee r\vdash q\vee r$)

Quantifiers: $\neg\forall x\,P(x)\equiv\exists x\,\neg P(x)$, $\;\neg\exists x\,P(x)\equiv\forall x\,\neg P(x)$

Normal forms: CNF (AND of ORs), DNF (OR of ANDs).


## 3. Proof techniques

Direct, contrapositive, contradiction, cases, mathematical induction (base + inductive step), strong induction, pigeonhole principle (if $n+1$ objects go into $n$ boxes, some box has $\ge 2$; generalized: some box has $\ge\lceil N/k\rceil$).


## 4. Combinatorics

Permutations $\frac{n!}{(n-r)!}$; combinations $\binom nr$; with repetition: $n^r$ (ordered), $\binom{n+r-1}{r}$ (unordered, stars and bars)

Permutations of a multiset: $\frac{n!}{n_1!n_2!\cdots n_k!}$; circular permutations $(n-1)!$

**Derangements:** $D_n = n!\sum_{k=0}^n\frac{(-1)^k}{k!}\approx\frac{n!}{e}$

**Catalan numbers:** $C_n = \frac{1}{n+1}\binom{2n}{n}$ (binary trees, valid parentheses)

**Stirling numbers of the second kind** $S(n,k)$: partitions of $n$ items into $k$ non-empty sets; **Bell numbers** $B_n = \sum_kS(n,k)$

**Vandermonde's identity:** $\binom{m+n}{r} = \sum_k\binom mk\binom n{r-k}$


## 5. Recurrence relations and generating functions

Linear homogeneous $a_n = c_1a_{n-1} + c_2a_{n-2}$: characteristic equation $r^2 = c_1r + c_2$; distinct roots $a_n = \alpha r_1^n + \beta r_2^n$; repeated $a_n = (\alpha + \beta n)r^n$

**Fibonacci:** $F_n = F_{n-1} + F_{n-2}$, $\;F_n = \frac{\varphi^n - \psi^n}{\sqrt5}$, $\varphi = \frac{1+\sqrt5}2$

**Generating function:** $G(x) = \sum a_nx^n$; e.g. $\frac1{1-x}\leftrightarrow1,1,1,\dots$; $\frac1{(1-x)^k}\leftrightarrow\binom{n+k-1}{n}$

**Master theorem** for $T(n) = aT(n/b) + f(n)$: compare $f(n)$ with $n^{\log_ba}$:
- $f = O(n^{\log_ba-\epsilon})$ ⇒ $T = \Theta(n^{\log_ba})$
- $f = \Theta(n^{\log_ba})$ ⇒ $T = \Theta(n^{\log_ba}\log n)$
- $f = \Omega(n^{\log_ba+\epsilon})$ (plus regularity) ⇒ $T = \Theta(f(n))$


## 6. Asymptotic notation

$f = O(g)$: $f\le cg$ eventually; $\Omega$: $\ge$; $\Theta$: both. Common orders: $1 < \log n < \sqrt n < n < n\log n < n^2 < 2^n < n!$

Self-attention is $O(n^2d)$ in sequence length $n$; linear attention / SSMs are $O(nd)$.


## 7. Graph theory basics

**Handshaking lemma:** $\sum_v\deg(v) = 2|E|$ (number of odd-degree vertices is even)

Complete graph $K_n$ has $\binom n2$ edges; a tree on $n$ vertices has $n-1$ edges; Cayley's formula: $n^{n-2}$ labelled trees on $n$ vertices

**Euler's formula (planar, connected):** $V - E + F = 2$; for simple planar graphs $E\le3V - 6$

**Euler circuit** exists iff connected and every vertex has even degree; Euler path iff exactly 0 or 2 odd-degree vertices. **Hamiltonian cycle:** Dirac's condition $\deg(v)\ge n/2$ is sufficient.

Bipartite iff no odd cycle; **König's theorem:** in bipartite graphs, max matching = min vertex cover. **Hall's marriage theorem.**

Chromatic number $\chi(G)$; four colour theorem for planar graphs; greedy colouring uses at most $\Delta + 1$ colours.


## 8. Graph matrices

Adjacency $A$; degree $D = \text{diag}(\sum_jA_{ij})$; $(A^k)_{ij}$ = number of walks of length $k$ from $i$ to $j$

**Graph Laplacian:** $L = D - A$; normalized $L_{sym} = I - D^{-1/2}AD^{-1/2}$; random-walk $L_{rw} = I - D^{-1}A$

$x^\top Lx = \sum_{(i,j)\in E}(x_i - x_j)^2\ge0$; the number of zero eigenvalues of $L$ = number of connected components; the second-smallest eigenvalue (Fiedler value) measures connectivity

**Matrix-tree theorem:** number of spanning trees = any cofactor of $L$

**ML note:** GCN layers use $\tilde D^{-1/2}\tilde A\tilde D^{-1/2}H W$ with $\tilde A = A + I$; spectral clustering uses the eigenvectors of $L$.


## 9. Graph algorithms (with complexity)

| Problem | Algorithm | Complexity |
|---|---|---|
| Traversal | BFS, DFS | $O(V+E)$ |
| Shortest path, non-negative weights | Dijkstra (binary heap) | $O((V+E)\log V)$ |
| Shortest path, negative weights | Bellman-Ford | $O(VE)$ |
| All-pairs shortest paths | Floyd-Warshall | $O(V^3)$ |
| Minimum spanning tree | Kruskal, Prim | $O(E\log V)$ |
| Topological sort | Kahn / DFS | $O(V+E)$ |
| Strongly connected components | Tarjan, Kosaraju | $O(V+E)$ |
| Max flow | Ford-Fulkerson, Edmonds-Karp | $O(VE^2)$ |
| Bipartite matching | Hopcroft-Karp | $O(E\sqrt V)$ |
| Ranking | PageRank $r = \frac{1-d}{N}\mathbf 1 + dA^\top D^{-1}r$ | iterative |

**Max-flow min-cut theorem:** the maximum flow equals the capacity of the minimum cut.


## 10. Algebraic structures

Group (closure, associativity, identity, inverses), abelian group, ring, field. Lagrange's theorem: the order of a subgroup divides the order of the group.

**Modular arithmetic:** Fermat's little theorem $a^{p-1}\equiv1\pmod p$; Euler $a^{\varphi(n)}\equiv1\pmod n$; Chinese remainder theorem; RSA uses $ed\equiv1\pmod{\varphi(n)}$.

**Boolean algebra:** $x + xy = x$, $x(x+y) = x$, $x + \bar xy = x + y$; Karnaugh maps.

**ML note:** group theory is the language of equivariant networks (CNNs are translation-equivariant; E(n)-GNNs are rotation-equivariant).
