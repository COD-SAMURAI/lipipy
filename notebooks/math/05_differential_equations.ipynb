# 05 · Ordinary and Partial Differential Equations

- **Level:** Fundamentals → Intermediate
- **Typical semester:** Semester 1–3 (first-order ODEs in Maths I, higher-order ODEs in Maths II, PDEs in Maths III)
- **Prerequisites:** 02, 03, 04
- **Used later in:** gradient flow, neural ODEs, diffusion models (SDEs), physics-informed networks, state-space models.


## 1. First-order ODEs

**Variable separable:** $\frac{dy}{dx} = f(x)g(y) \Rightarrow \int\frac{dy}{g(y)} = \int f(x)\,dx$

**Homogeneous:** $\frac{dy}{dx} = F\left(\frac yx\right)$; substitute $y = vx$, $\frac{dy}{dx} = v + x\frac{dv}{dx}$.

**Linear (Leibniz) equation:** $\frac{dy}{dx} + P(x)y = Q(x)$. Integrating factor $\mu = e^{\int P\,dx}$, solution $y\mu = \int Q\mu\,dx + C$.

**Bernoulli equation:** $\frac{dy}{dx} + Py = Qy^n$; substitute $z = y^{1-n}$ to make it linear.

**Exact equation:** $M\,dx + N\,dy = 0$ is exact if $\frac{\partial M}{\partial y} = \frac{\partial N}{\partial x}$; solution $\int M\,dx\,(y\text{ const}) + \int(\text{terms of }N\text{ free of }x)\,dy = C$.

**Integrating factors for non-exact:** if $\frac{M_y - N_x}{N} = f(x)$ use $e^{\int f\,dx}$; if $\frac{N_x - M_y}{M} = g(y)$ use $e^{\int g\,dy}$.

**Clairaut's equation:** $y = px + f(p)$, $p = y'$; general solution $y = cx + f(c)$.


## 2. Applications of first-order ODEs

**Exponential growth/decay:** $\frac{dN}{dt} = kN \Rightarrow N = N_0e^{kt}$
**Newton's law of cooling:** $\frac{dT}{dt} = -k(T - T_s)$
**RC / RL circuits:** $L\frac{di}{dt} + Ri = E$
**Orthogonal trajectories:** replace $\frac{dy}{dx}$ by $-\frac{dx}{dy}$ (Cartesian) or $r\frac{d\theta}{dr}$ by $-\frac{1}{r}\frac{dr}{d\theta}$ (polar)
**Logistic growth:** $\frac{dN}{dt} = rN\left(1-\frac NK\right) \Rightarrow N(t) = \frac{K}{1+Ae^{-rt}}$, which is a scaled sigmoid


## 3. Higher-order linear ODEs with constant coefficients

$a_ny^{(n)} + \cdots + a_1y' + a_0y = f(x)$. Solution $y = y_c + y_p$.

**Complementary function** from the auxiliary equation $a_nm^n + \cdots + a_0 = 0$:
- distinct real roots $m_1, m_2$: $y_c = c_1e^{m_1x} + c_2e^{m_2x}$
- repeated root $m$: $y_c = (c_1 + c_2x)e^{mx}$
- complex roots $\alpha\pm i\beta$: $y_c = e^{\alpha x}(c_1\cos\beta x + c_2\sin\beta x)$

**Particular integral** with operator $D = \frac{d}{dx}$, $y_p = \frac{1}{F(D)}f(x)$:
- $f = e^{ax}$: $y_p = \frac{e^{ax}}{F(a)}$ if $F(a)\ne0$; otherwise $\frac{xe^{ax}}{F'(a)}$
- $f = \sin ax$ or $\cos ax$: replace $D^2$ by $-a^2$
- $f = x^n$: expand $[F(D)]^{-1}$ as a power series in $D$
- $f = e^{ax}V(x)$: $\frac{1}{F(D)}e^{ax}V = e^{ax}\frac{1}{F(D+a)}V$

**Method of variation of parameters:** for $y'' + py' + qy = f$ with $y_c = c_1y_1 + c_2y_2$ and Wronskian $W = y_1y_2' - y_2y_1'$:
$$y_p = -y_1\int\frac{y_2f}{W}dx + y_2\int\frac{y_1f}{W}dx$$

**Cauchy-Euler equation:** $x^2y'' + axy' + by = f(x)$; substitute $x = e^t$, $xD = \theta$, $x^2D^2 = \theta(\theta-1)$.

**Legendre's equation:** $(ax+b)^2y'' + \cdots$; substitute $ax + b = e^t$.


## 4. Mechanical and electrical oscillations

$$m\ddot x + c\dot x + kx = F(t), \qquad L\ddot q + R\dot q + \frac{q}{C} = E(t)$$

Natural frequency $\omega_0 = \sqrt{k/m}$, damping ratio $\zeta = \frac{c}{2\sqrt{mk}}$: overdamped $\zeta>1$, critically damped $\zeta=1$, underdamped $\zeta<1$.

**ML note:** momentum SGD on a quadratic is exactly a discretized damped oscillator.


## 5. Series solutions and special functions

**Frobenius method:** near a regular singular point try $y = \sum_{n=0}^\infty a_nx^{n+r}$; the indicial equation gives $r$.

**Bessel's equation:** $x^2y'' + xy' + (x^2-n^2)y = 0$, $\quad J_n(x) = \sum_{k=0}^\infty\frac{(-1)^k}{k!\,\Gamma(n+k+1)}\left(\frac x2\right)^{n+2k}$

**Legendre's equation:** $(1-x^2)y'' - 2xy' + n(n+1)y = 0$; Rodrigues' formula $P_n(x) = \frac{1}{2^nn!}\frac{d^n}{dx^n}(x^2-1)^n$; orthogonality $\int_{-1}^1P_mP_n\,dx = \frac{2}{2n+1}\delta_{mn}$

**Hermite, Laguerre, Chebyshev polynomials** appear in quantum mechanics, Gaussian quadrature and the HiPPO / S4 state-space models (Legendre and Laguerre bases).


## 6. Systems of ODEs

$\dot{\mathbf x} = A\mathbf x \Rightarrow \mathbf x(t) = e^{At}\mathbf x_0$. If $A = V\Lambda V^{-1}$, $\;e^{At} = Ve^{\Lambda t}V^{-1}$.

Stability: the origin is asymptotically stable iff all eigenvalues of $A$ have negative real part.

**ML note:** linear state-space layers (S4, Mamba) use $\dot h = Ah + Bu$, $y = Ch$ and discretize it (see 24 · Sequence models).


## 7. Partial differential equations: formation and first order

Form a PDE by eliminating arbitrary constants or functions. Notation $p = z_x$, $q = z_y$, $r = z_{xx}$, $s = z_{xy}$, $t = z_{yy}$.

**Lagrange's linear equation:** $Pp + Qq = R$; auxiliary equations $\frac{dx}{P} = \frac{dy}{Q} = \frac{dz}{R}$, solution $\phi(u,v) = 0$ from two independent integrals $u = c_1$, $v = c_2$.

**Charpit's method** for nonlinear $f(x,y,z,p,q) = 0$: $\frac{dx}{f_p} = \frac{dy}{f_q} = \frac{dz}{pf_p+qf_q} = \frac{-dp}{f_x+pf_z} = \frac{-dq}{f_y+qf_z}$.


## 8. Second-order PDEs

Classification of $Au_{xx} + Bu_{xy} + Cu_{yy} + \cdots = 0$: elliptic $B^2-4AC<0$, parabolic $=0$, hyperbolic $>0$.

**One-dimensional heat equation (parabolic):** $u_t = c^2u_{xx}$. With $u(0,t) = u(L,t) = 0$:
$$u(x,t) = \sum_{n=1}^\infty b_n\sin\frac{n\pi x}{L}e^{-c^2n^2\pi^2t/L^2}$$

**One-dimensional wave equation (hyperbolic):** $u_{tt} = c^2u_{xx}$.
$$u(x,t) = \sum_{n=1}^\infty\left(a_n\cos\frac{n\pi ct}{L} + b_n\sin\frac{n\pi ct}L\right)\sin\frac{n\pi x}{L}$$
D'Alembert's solution: $u = \frac12[f(x+ct) + f(x-ct)] + \frac{1}{2c}\int_{x-ct}^{x+ct}g(s)\,ds$.

**Laplace's equation (elliptic):** $\nabla^2u = u_{xx} + u_{yy} = 0$; in polar form $u_{rr} + \frac1ru_r + \frac{1}{r^2}u_{\theta\theta} = 0$.

**Method:** separation of variables $u = X(x)T(t)$ turns a PDE into ODEs.

**ML note:** the heat equation is the forward process of diffusion models (adding Gaussian noise), and the Fokker-Planck equation $\partial_tp = -\nabla\cdot(fp) + \frac{g^2}{2}\nabla^2p$ describes how the noise evolves.


## 9. Stochastic differential equations (bridge to modern ML)

$$dX_t = f(X_t,t)\,dt + g(t)\,dW_t$$

**Itô's lemma:** $d\phi(X_t) = \left(\phi'f + \frac12\phi''g^2\right)dt + \phi'g\,dW_t$

**Ornstein-Uhlenbeck process:** $dX_t = -\theta X_t\,dt + \sigma\,dW_t$ (the "variance preserving" SDE of diffusion models).

**Reverse-time SDE (Anderson):** $dX_t = \left[f - g^2\nabla_x\log p_t(X_t)\right]dt + g\,d\bar W_t$, which is how score-based diffusion models generate samples (see 20 · Generative models).
