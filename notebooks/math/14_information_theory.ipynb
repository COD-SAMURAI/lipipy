# 14 · Information Theory

- **Level:** Intermediate
- **Typical semester:** Semester 5–6 (Information Theory and Coding, or Digital Communication)
- **Prerequisites:** 11
- **Used later in:** cross-entropy loss, KL terms in VAEs and RLHF, mutual information in contrastive learning, compression and tokenization, perplexity.


## 1. Self-information and entropy

Self-information of an outcome: $I(x) = -\log p(x)$ (bits for $\log_2$, nats for $\ln$)

**Shannon entropy:** $H(X) = -\sum_xp(x)\log p(x) = \mathbb E[-\log p(X)]$; $\;0\le H(X)\le\log|\mathcal X|$, maximum for the uniform distribution

Binary entropy: $H_b(p) = -p\log p - (1-p)\log(1-p)$

**Differential entropy:** $h(X) = -\int f\log f$; Gaussian $h = \frac12\log(2\pi e\sigma^2)$, multivariate $\frac12\log\left((2\pi e)^d|\Sigma|\right)$. The Gaussian has maximum entropy for a fixed variance.


## 2. Joint, conditional and mutual information

$H(X,Y) = -\sum p(x,y)\log p(x,y)$; $\;H(Y|X) = H(X,Y) - H(X)$ (chain rule)

**Mutual information:** $I(X;Y) = H(X) - H(X|Y) = H(X) + H(Y) - H(X,Y) = D_{KL}(p(x,y)\,\|\,p(x)p(y))\ge0$

Conditioning reduces entropy: $H(X|Y)\le H(X)$.

**Data processing inequality:** if $X\to Y\to Z$ is a Markov chain, $I(X;Z)\le I(X;Y)$.


## 3. Divergences between distributions

**Cross-entropy:** $H(p,q) = -\sum p\log q = H(p) + D_{KL}(p\|q)$

**Kullback-Leibler divergence:** $D_{KL}(p\|q) = \sum p\log\frac pq\ge0$ (Gibbs' inequality), not symmetric

KL between Gaussians:
$$D_{KL}(\mathcal N(\mu_1,\sigma_1^2)\|\mathcal N(\mu_2,\sigma_2^2)) = \log\frac{\sigma_2}{\sigma_1} + \frac{\sigma_1^2 + (\mu_1-\mu_2)^2}{2\sigma_2^2} - \frac12$$
$$D_{KL}(\mathcal N(\mu,\text{diag}\,\sigma^2)\|\mathcal N(0,I)) = \frac12\sum_j\left(\sigma_j^2 + \mu_j^2 - 1 - \log\sigma_j^2\right) \quad\text{(the VAE KL term)}$$

**Jensen-Shannon:** $JS(p\|q) = \frac12D_{KL}(p\|m) + \frac12D_{KL}(q\|m)$, $m = \frac{p+q}{2}$ (the original GAN objective)

**$f$-divergences:** $D_f(p\|q) = \mathbb E_q\left[f\left(\frac pq\right)\right]$; KL ($f = t\log t$), reverse KL, total variation ($f = \frac12|t-1|$), Hellinger, $\chi^2$

**Wasserstein-1 (earth mover's):** $W_1(p,q) = \inf_{\gamma\in\Pi(p,q)}\mathbb E_{(x,y)\sim\gamma}\|x - y\| = \sup_{\|f\|_L\le1}\mathbb E_p f - \mathbb E_qf$ (Kantorovich-Rubinstein duality; used in WGAN)

**Pinsker's inequality:** $TV(p,q)\le\sqrt{\frac12D_{KL}(p\|q)}$

**Rényi entropy:** $H_\alpha = \frac1{1-\alpha}\log\sum p^\alpha$ ($\alpha\to1$ gives Shannon)


## 4. Source coding and compression

**Kraft inequality:** a prefix code with lengths $\ell_i$ exists iff $\sum2^{-\ell_i}\le1$

**Shannon's source coding theorem:** the expected code length satisfies $H(X)\le\bar L < H(X) + 1$

Huffman coding (optimal prefix code), Shannon-Fano coding, arithmetic coding, Lempel-Ziv. Code efficiency $\eta = \frac{H}{\bar L}$, redundancy $1-\eta$.

**ML note:** a language model with cross-entropy $H$ bits/token can compress text to about $H$ bits/token with arithmetic coding, so "prediction = compression". Byte-pair encoding (BPE) is a compression algorithm reused as a tokenizer.


## 5. Channel capacity

$C = \max_{p(x)}I(X;Y)$. Binary symmetric channel with flip probability $p$: $C = 1 - H_b(p)$. Binary erasure channel: $C = 1-\epsilon$.

**Shannon-Hartley theorem:** $C = B\log_2\left(1 + \frac SN\right)$ bits/s

**Channel coding theorem:** reliable communication is possible at any rate $R<C$, and impossible for $R>C$.

Error-correcting codes: Hamming distance, minimum distance $d$ detects $d-1$ and corrects $\lfloor\frac{d-1}2\rfloor$ errors; Hamming bound; linear block codes $c = mG$, syndrome $s = rH^\top$; cyclic codes, convolutional codes (Viterbi decoding), LDPC, turbo and polar codes.


## 6. Rate-distortion theory

$R(D) = \min_{p(\hat x|x):\,\mathbb E d(X,\hat X)\le D}I(X;\hat X)$; Gaussian source with MSE: $R(D) = \frac12\log\frac{\sigma^2}{D}$

**ML note:** $\beta$-VAEs and learned image compression trade off rate (KL) and distortion (reconstruction).


## 7. Information theory in ML

**Perplexity:** $\text{PPL} = \exp\left(-\frac1N\sum_t\log p(x_t|x_{<t})\right) = e^{\text{cross-entropy in nats}}$

**Information bottleneck:** $\min_{p(z|x)}I(X;Z) - \beta I(Z;Y)$

**InfoNCE bound:** $I(X;Y)\ge\log N - \mathcal L_{\text{InfoNCE}}$ (contrastive learning maximizes a lower bound on mutual information)

**Minimum description length (MDL):** choose the model minimizing $L(\text{model}) + L(\text{data}|\text{model})$

**Kolmogorov complexity:** $K(x)$ = length of the shortest program that outputs $x$; uncomputable, but the theoretical ideal of compression (see 25 · Theory toward AGI).

**Fisher information** (from 12 §4) is the curvature of KL: $D_{KL}(p_\theta\|p_{\theta+d\theta})\approx\frac12d\theta^\top I(\theta)d\theta$.
