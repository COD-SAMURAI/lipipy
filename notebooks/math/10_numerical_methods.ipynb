# 10 · Numerical Methods

- **Level:** Intermediate
- **Typical semester:** Semester 3–4 (Numerical Methods, often part of Maths III/IV)
- **Prerequisites:** 02, 03, 05, 06
- **Used later in:** every training loop is a numerical method. Newton's method, ODE solvers for neural ODEs and diffusion samplers, iterative linear solvers.


## 1. Errors

Absolute error $|x - \tilde x|$, relative error $\frac{|x-\tilde x|}{|x|}$. Truncation error comes from approximating a formula; round-off error from finite precision.

Machine epsilon: float64 $\approx 2.2\times10^{-16}$, float32 $\approx 1.2\times10^{-7}$, bfloat16 $\approx 7.8\times10^{-3}$.

**Numerically stable log-sum-exp:** $\log\sum_ie^{x_i} = m + \log\sum_ie^{x_i - m}$, $m = \max_ix_i$.


## 2. Roots of equations $f(x) = 0$

**Bisection:** if $f(a)f(b)<0$, $c = \frac{a+b}2$, keep the half with the sign change. Error after $n$ steps $\le\frac{b-a}{2^n}$ (linear convergence).

**Regula falsi (false position):** $x = \frac{af(b) - bf(a)}{f(b) - f(a)}$

**Secant:** $x_{n+1} = x_n - f(x_n)\frac{x_n - x_{n-1}}{f(x_n) - f(x_{n-1})}$ (order $\approx1.618$)

**Newton-Raphson:** $x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$ (quadratic convergence near a simple root)
- square root of $N$: $x_{n+1} = \frac12\left(x_n + \frac{N}{x_n}\right)$; reciprocal: $x_{n+1} = x_n(2 - Nx_n)$

**Fixed-point iteration:** $x_{n+1} = g(x_n)$ converges if $|g'(x)|<1$ near the root.

**Newton for systems:** $\mathbf x_{n+1} = \mathbf x_n - J(\mathbf x_n)^{-1}\mathbf F(\mathbf x_n)$


## 3. Linear systems

**Direct:** Gauss elimination (with partial pivoting), Gauss-Jordan, LU (Doolittle $L_{ii}=1$, Crout $U_{ii}=1$), Cholesky for SPD. Cost $O(n^3)$.

**Iterative:**
- Jacobi: $x_i^{(k+1)} = \frac1{a_{ii}}\left(b_i - \sum_{j\ne i}a_{ij}x_j^{(k)}\right)$
- Gauss-Seidel: use the newest values $x_j^{(k+1)}$ for $j<i$
- Successive over-relaxation (SOR): $x^{(k+1)} = (1-\omega)x^{(k)} + \omega x_{GS}^{(k+1)}$
- Converge if $A$ is strictly diagonally dominant: $|a_{ii}| > \sum_{j\ne i}|a_{ij}|$
- Conjugate gradient (SPD $A$): converges in at most $n$ steps exactly; rate depends on $\sqrt{\kappa}$

**Eigenvalues:** power method, inverse power method, Rayleigh quotient iteration, QR algorithm, Lanczos / Arnoldi for large sparse matrices.


## 4. Finite differences and interpolation

Forward $\Delta f_i = f_{i+1} - f_i$, backward $\nabla f_i = f_i - f_{i-1}$, central $\delta f_i = f_{i+1/2} - f_{i-1/2}$, shift $Ef(x) = f(x+h)$, with $\Delta = E - 1$ and $E = e^{hD}$.

**Newton forward interpolation** ($u = \frac{x-x_0}{h}$):
$$f(x) = f_0 + u\Delta f_0 + \frac{u(u-1)}{2!}\Delta^2f_0 + \frac{u(u-1)(u-2)}{3!}\Delta^3f_0 + \cdots$$

**Newton backward** ($u = \frac{x - x_n}h$): $f(x) = f_n + u\nabla f_n + \frac{u(u+1)}{2!}\nabla^2f_n + \cdots$

**Lagrange interpolation:** $P(x) = \sum_iy_i\prod_{j\ne i}\frac{x - x_j}{x_i - x_j}$

**Newton divided differences:** $f[x_0,x_1] = \frac{f_1 - f_0}{x_1 - x_0}$, $P(x) = f_0 + (x - x_0)f[x_0,x_1] + (x-x_0)(x-x_1)f[x_0,x_1,x_2] + \cdots$

Also: Gauss forward/backward, Stirling, Bessel (central difference formulas), cubic splines.


## 5. Numerical differentiation

$f'(x)\approx\frac{f(x+h)-f(x)}{h}$ (error $O(h)$), $\;\frac{f(x+h)-f(x-h)}{2h}$ (error $O(h^2)$), $\;f''(x)\approx\frac{f(x+h)-2f(x)+f(x-h)}{h^2}$


## 6. Numerical integration (quadrature)

With $h = \frac{b-a}{n}$, $y_i = f(a+ih)$:

**Trapezoidal rule:** $\int_a^bf\,dx\approx\frac h2[y_0 + 2(y_1 + \cdots + y_{n-1}) + y_n]$, error $O(h^2)$

**Simpson's 1/3 rule** ($n$ even): $\frac h3[y_0 + 4(y_1 + y_3 + \cdots) + 2(y_2 + y_4 + \cdots) + y_n]$, error $O(h^4)$

**Simpson's 3/8 rule** ($n$ multiple of 3): $\frac{3h}8[y_0 + 3(y_1+y_2+y_4+y_5+\cdots) + 2(y_3+y_6+\cdots) + y_n]$

**Weddle's rule** ($n$ multiple of 6): $\frac{3h}{10}[y_0 + 5y_1 + y_2 + 6y_3 + y_4 + 5y_5 + y_6]$ per block

**Boole's rule**, **Romberg integration** (Richardson extrapolation of trapezoid), **Gaussian quadrature** ($n$ points exact for polynomials of degree $2n-1$)

**Monte Carlo integration:** $\int f(x)p(x)\,dx\approx\frac1N\sum f(x_i)$, $x_i\sim p$, error $O(1/\sqrt N)$ regardless of dimension.


## 7. Numerical solution of ODEs $y' = f(x,y)$

**Taylor series method:** $y_{n+1} = y_n + hy_n' + \frac{h^2}{2!}y_n'' + \cdots$

**Picard's method:** $y^{(k+1)}(x) = y_0 + \int_{x_0}^xf(t, y^{(k)}(t))\,dt$

**Euler:** $y_{n+1} = y_n + hf(x_n,y_n)$ (first order)

**Modified Euler / Heun:** $y_{n+1} = y_n + \frac h2[f(x_n,y_n) + f(x_{n+1}, y_n + hf(x_n,y_n))]$

**Runge-Kutta 4th order:**
$$k_1 = hf(x_n,y_n),\; k_2 = hf\!\left(x_n+\tfrac h2, y_n+\tfrac{k_1}2\right),\; k_3 = hf\!\left(x_n+\tfrac h2, y_n+\tfrac{k_2}{2}\right),\; k_4 = hf(x_n+h, y_n+k_3)$$
$$y_{n+1} = y_n + \frac16(k_1 + 2k_2 + 2k_3 + k_4)$$

**Predictor-corrector:** Milne's method, Adams-Bashforth-Moulton.

**Stiff equations:** implicit (backward) Euler $y_{n+1} = y_n + hf(x_{n+1},y_{n+1})$.

**ML note:** gradient descent is Euler's method on the gradient flow $\dot\theta = -\nabla L(\theta)$. Diffusion samplers (DDIM, DPM-Solver, Heun in EDM) are ODE solvers.


## 8. Numerical PDEs

Finite-difference grids: $u_{xx}\approx\frac{u_{i+1,j} - 2u_{i,j} + u_{i-1,j}}{h^2}$.

Laplace equation: standard five-point formula $u_{i,j} = \frac14(u_{i+1,j} + u_{i-1,j} + u_{i,j+1} + u_{i,j-1})$ (Liebmann iteration).

Heat equation: explicit Bender-Schmidt scheme (stable if $r = \frac{c^2k}{h^2}\le\frac12$), implicit Crank-Nicolson (unconditionally stable).

Wave equation: explicit scheme stable if the Courant number $\frac{ck}{h}\le1$ (CFL condition).


## 9. Curve fitting (least squares)

Fit $y = a + bx$ by the normal equations $\sum y = na + b\sum x$, $\;\sum xy = a\sum x + b\sum x^2$.

Fit $y = ae^{bx}$ by taking logs: $\ln y = \ln a + bx$. Fit $y = ax^b$: $\log y = \log a + b\log x$.

General: minimize $\|Ax - b\|^2\Rightarrow A^\top Ax = A^\top b$ (see 17 · Classic ML §1).
