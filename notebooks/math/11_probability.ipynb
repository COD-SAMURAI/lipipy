# 11 · Probability Theory

- **Level:** Fundamentals → Intermediate
- **Typical semester:** Semester 3–4 (Probability and Statistics)
- **Prerequisites:** 01, 03
- **Used later in:** every probabilistic model, loss functions as negative log-likelihoods, generative models, Bayesian methods, RL.


## 1. Axioms and basic rules

$P(\Omega) = 1$, $P(A)\ge0$, and for disjoint events $P(\cup A_i) = \sum P(A_i)$.

$P(A^c) = 1 - P(A)$, $\;P(A\cup B) = P(A) + P(B) - P(A\cap B)$

Inclusion-exclusion: $P(A\cup B\cup C) = \sum P(A) - \sum P(A\cap B) + P(A\cap B\cap C)$


## 2. Conditional probability, independence, Bayes

$$P(A|B) = \frac{P(A\cap B)}{P(B)}, \qquad P(A\cap B) = P(A|B)P(B)$$

Independence: $P(A\cap B) = P(A)P(B)$. Conditional independence: $P(A\cap B|C) = P(A|C)P(B|C)$.

**Total probability:** $P(B) = \sum_iP(B|A_i)P(A_i)$

**Bayes' theorem:**
$$P(A_i|B) = \frac{P(B|A_i)P(A_i)}{\sum_jP(B|A_j)P(A_j)}, \qquad \text{posterior}\propto\text{likelihood}\times\text{prior}$$

**Chain rule of probability:** $p(x_1,\dots,x_n) = \prod_{t=1}^np(x_t|x_{<t})$, which is exactly how autoregressive language models factorize text.


## 3. Random variables

Discrete: PMF $p(x)$, $\sum p(x) = 1$. Continuous: PDF $f(x)$, $\int f = 1$, CDF $F(x) = P(X\le x)$, $f = F'$.

**Expectation:** $\mathbb E[X] = \sum xp(x)$ or $\int xf(x)dx$; $\;\mathbb E[g(X)] = \int g(x)f(x)dx$ (LOTUS); linearity $\mathbb E[aX+bY] = a\mathbb E X + b\mathbb EY$

**Variance:** $\text{Var}(X) = \mathbb E[(X-\mu)^2] = \mathbb E[X^2] - \mu^2$, $\;\text{Var}(aX+b) = a^2\text{Var}X$

**Covariance:** $\text{Cov}(X,Y) = \mathbb E[XY] - \mathbb EX\,\mathbb EY$; correlation $\rho = \frac{\text{Cov}}{\sigma_X\sigma_Y}\in[-1,1]$

$\text{Var}(X+Y) = \text{Var}X + \text{Var}Y + 2\text{Cov}(X,Y)$

**Moments:** $\mu_r' = \mathbb E[X^r]$; skewness $\frac{\mathbb E(X-\mu)^3}{\sigma^3}$; kurtosis $\frac{\mathbb E(X-\mu)^4}{\sigma^4}$

**Moment generating function:** $M_X(t) = \mathbb E[e^{tX}]$, $\mathbb E[X^r] = M^{(r)}(0)$. Characteristic function $\varphi(t) = \mathbb E[e^{itX}]$.

**Law of total expectation / variance:** $\mathbb E[X] = \mathbb E[\mathbb E[X|Y]]$, $\;\text{Var}X = \mathbb E[\text{Var}(X|Y)] + \text{Var}(\mathbb E[X|Y])$


## 4. Discrete distributions

| Distribution | PMF | Mean | Variance |
|---|---|---|---|
| Bernoulli$(p)$ | $p^x(1-p)^{1-x}$ | $p$ | $p(1-p)$ |
| Binomial$(n,p)$ | $\binom nxp^x(1-p)^{n-x}$ | $np$ | $np(1-p)$ |
| Poisson$(\lambda)$ | $\frac{e^{-\lambda}\lambda^x}{x!}$ | $\lambda$ | $\lambda$ |
| Geometric$(p)$ | $(1-p)^{x-1}p$, $x\ge1$ | $\frac1p$ | $\frac{1-p}{p^2}$ |
| Negative binomial$(r,p)$ | $\binom{x-1}{r-1}p^r(1-p)^{x-r}$ | $\frac rp$ | $\frac{r(1-p)}{p^2}$ |
| Hypergeometric | $\frac{\binom Kx\binom{N-K}{n-x}}{\binom Nn}$ | $n\frac KN$ | $n\frac KN\frac{N-K}N\frac{N-n}{N-1}$ |
| Categorical$(\pi)$ | $\prod_k\pi_k^{[x=k]}$ | | |
| Multinomial | $\frac{n!}{\prod x_k!}\prod\pi_k^{x_k}$ | $n\pi_k$ | $n\pi_k(1-\pi_k)$ |

Poisson approximation to binomial: $n\to\infty$, $p\to0$, $np = \lambda$.


## 5. Continuous distributions

| Distribution | PDF | Mean | Variance |
|---|---|---|---|
| Uniform$(a,b)$ | $\frac1{b-a}$ | $\frac{a+b}2$ | $\frac{(b-a)^2}{12}$ |
| Exponential$(\lambda)$ | $\lambda e^{-\lambda x}$ | $\frac1\lambda$ | $\frac1{\lambda^2}$ |
| Normal$(\mu,\sigma^2)$ | $\frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{(x-\mu)^2}{2\sigma^2}}$ | $\mu$ | $\sigma^2$ |
| Gamma$(\alpha,\beta)$ | $\frac{\beta^\alpha}{\Gamma(\alpha)}x^{\alpha-1}e^{-\beta x}$ | $\frac\alpha\beta$ | $\frac\alpha{\beta^2}$ |
| Beta$(\alpha,\beta)$ | $\frac{x^{\alpha-1}(1-x)^{\beta-1}}{B(\alpha,\beta)}$ | $\frac{\alpha}{\alpha+\beta}$ | $\frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}$ |
| Laplace$(\mu,b)$ | $\frac1{2b}e^{-|x-\mu|/b}$ | $\mu$ | $2b^2$ |
| Student-$t_\nu$ | $\propto\left(1+\frac{x^2}{\nu}\right)^{-\frac{\nu+1}2}$ | $0$ | $\frac{\nu}{\nu-2}$ |
| Chi-square $\chi^2_k$ | Gamma$(k/2, 1/2)$ | $k$ | $2k$ |
| Log-normal | $\frac{1}{x\sigma\sqrt{2\pi}}e^{-\frac{(\ln x-\mu)^2}{2\sigma^2}}$ | $e^{\mu+\sigma^2/2}$ | $(e^{\sigma^2}-1)e^{2\mu+\sigma^2}$ |
| Weibull, Cauchy, Pareto, Gumbel | (heavy tails / extremes) | | |

Exponential is memoryless: $P(X>s+t|X>s) = P(X>t)$.

**Gumbel-max trick:** $\arg\max_k(\log\pi_k + g_k)$, $g_k\sim\text{Gumbel}(0,1)$, samples from Categorical$(\pi)$. Its smooth relaxation (Gumbel-softmax) is used in discrete latent models.


## 6. Multivariate Gaussian

$$\mathcal N(x|\mu,\Sigma) = \frac{1}{(2\pi)^{d/2}|\Sigma|^{1/2}}\exp\left(-\frac12(x-\mu)^\top\Sigma^{-1}(x-\mu)\right)$$

For partition $x = (x_a, x_b)$:
$$\mu_{a|b} = \mu_a + \Sigma_{ab}\Sigma_{bb}^{-1}(x_b - \mu_b), \qquad \Sigma_{a|b} = \Sigma_{aa} - \Sigma_{ab}\Sigma_{bb}^{-1}\Sigma_{ba}$$

Affine transform: $Ax + b\sim\mathcal N(A\mu + b, A\Sigma A^\top)$. Sampling: $x = \mu + L\epsilon$, $LL^\top = \Sigma$, $\epsilon\sim\mathcal N(0,I)$ (the reparameterization trick).

Product of Gaussians is (proportional to) a Gaussian; sum of independent Gaussians is Gaussian with added means and variances.


## 7. Joint distributions and transformations

Marginal $f_X(x) = \int f(x,y)\,dy$; conditional $f(y|x) = \frac{f(x,y)}{f_X(x)}$.

**Change of variables:** if $Y = g(X)$ invertible, $f_Y(y) = f_X(g^{-1}(y))\left|\frac{d g^{-1}}{dy}\right|$; multivariate $p_Y(y) = p_X(x)\left|\det\frac{\partial x}{\partial y}\right|$

Sum of independent variables: PDF is the convolution $f_{X+Y} = f_X * f_Y$, and $M_{X+Y} = M_XM_Y$.


## 8. Inequalities and limit theorems

**Markov:** $P(X\ge a)\le\frac{\mathbb EX}{a}$ ($X\ge0$). **Chebyshev:** $P(|X-\mu|\ge k\sigma)\le\frac1{k^2}$

**Jensen:** $f(\mathbb EX)\le\mathbb Ef(X)$ for convex $f$ (proves KL $\ge 0$ and the ELBO)

**Hoeffding:** for i.i.d. $X_i\in[a,b]$: $P(|\bar X - \mu|\ge t)\le2\exp\left(-\frac{2nt^2}{(b-a)^2}\right)$

**Chernoff bound:** $P(X\ge a)\le\min_{t>0}e^{-ta}M_X(t)$

**Union bound:** $P(\cup A_i)\le\sum P(A_i)$

**Law of large numbers:** $\bar X_n\to\mu$. **Central limit theorem:** $\frac{\bar X_n - \mu}{\sigma/\sqrt n}\xrightarrow{d}\mathcal N(0,1)$


## 9. Stochastic processes

**Markov chain:** $P(X_{t+1}|X_t,\dots,X_0) = P(X_{t+1}|X_t)$, transition matrix $P$, $\pi^{(t+1)} = \pi^{(t)}P$.
Stationary distribution $\pi = \pi P$. Detailed balance $\pi_iP_{ij} = \pi_jP_{ji}$ ⇒ stationarity (basis of MCMC).
Chapman-Kolmogorov: $P^{(m+n)} = P^{(m)}P^{(n)}$.

**Poisson process:** counts in time $t$ are Poisson$(\lambda t)$; inter-arrival times Exponential$(\lambda)$.

**Queueing (M/M/1):** utilization $\rho = \lambda/\mu$, mean number in system $L = \frac{\rho}{1-\rho}$, Little's law $L = \lambda W$.

**Brownian motion / Wiener process:** $W_t - W_s\sim\mathcal N(0,t-s)$, independent increments (see 05 §9).

**Gaussian process:** $f\sim\mathcal{GP}(m(x), k(x,x'))$, any finite set of values is jointly Gaussian.


## 10. Sampling methods

**Inverse transform:** $X = F^{-1}(U)$, $U\sim\text{Uniform}(0,1)$. **Box-Muller:** $Z = \sqrt{-2\ln U_1}\cos(2\pi U_2)$

**Rejection sampling:** accept $x\sim q$ with probability $\frac{p(x)}{Mq(x)}$

**Importance sampling:** $\mathbb E_p[f] = \mathbb E_q\left[f\frac pq\right]$; self-normalized weights $w_i\propto p(x_i)/q(x_i)$

**Metropolis-Hastings:** accept $x'\sim q(\cdot|x)$ with $\alpha = \min\left(1, \frac{p(x')q(x|x')}{p(x)q(x'|x)}\right)$

**Gibbs:** sample each variable from its full conditional $p(x_i|x_{-i})$

**Langevin dynamics:** $x_{t+1} = x_t + \frac\epsilon2\nabla\log p(x_t) + \sqrt\epsilon\,z_t$ (used by score-based generative models)

**Hamiltonian Monte Carlo:** simulate $H(q,p) = -\log\pi(q) + \frac12p^\top M^{-1}p$ with leapfrog steps.
