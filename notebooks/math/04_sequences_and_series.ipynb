# 04 · Sequences, Series and Fourier Series

- **Level:** Fundamentals
- **Typical semester:** Semester 1–3 (infinite series in Maths I/II, Fourier series in Maths III)
- **Prerequisites:** 02, 03
- **Used later in:** convergence of learning-rate schedules, power-series approximations, Fourier features and position encodings, spectral methods.


## 1. Sequences and convergence

A sequence $a_n \to L$ if for every $\varepsilon>0$ there is $N$ with $|a_n - L|<\varepsilon$ for all $n>N$.

Monotone convergence: a bounded monotone sequence converges. Standard limits: $\lim\sqrt[n]{n} = 1$, $\lim\frac{n!}{n^n} = 0$, $\lim x^n = 0$ for $|x|<1$.


## 2. Tests for convergence of $\sum a_n$

**$n$-th term test:** if $a_n \not\to 0$ the series diverges.

**$p$-series:** $\sum\frac{1}{n^p}$ converges iff $p>1$ (harmonic series $p=1$ diverges).

**Comparison:** $0\le a_n\le b_n$ and $\sum b_n$ converges ⇒ $\sum a_n$ converges. **Limit comparison:** $\lim\frac{a_n}{b_n} = c\in(0,\infty)$ ⇒ both behave the same.

**D'Alembert's ratio test:** $L = \lim\left|\frac{a_{n+1}}{a_n}\right|$; converges if $L<1$, diverges if $L>1$.

**Cauchy's root test:** $L = \lim\sqrt[n]{|a_n|}$; same conclusion.

**Raabe's test:** $\lim n\left(\frac{a_n}{a_{n+1}}-1\right) > 1$ ⇒ converges, $<1$ ⇒ diverges.

**Integral test:** for positive decreasing $f$, $\sum f(n)$ and $\int_1^\infty f(x)dx$ converge or diverge together.

**Leibniz (alternating series):** $\sum(-1)^nb_n$ converges if $b_n$ decreases to 0.

**Absolute vs conditional convergence:** $\sum|a_n|$ converges ⇒ $\sum a_n$ converges (absolutely).

**ML note:** Robbins-Monro conditions for SGD step sizes: $\sum_t\eta_t = \infty$, $\sum_t\eta_t^2 < \infty$ (e.g. $\eta_t = 1/t$).


## 3. Power series

$\sum c_n(x-a)^n$ has radius of convergence $R = \frac{1}{\limsup\sqrt[n]{|c_n|}} = \lim\left|\frac{c_n}{c_{n+1}}\right|$.

Inside the radius, a power series can be differentiated and integrated term by term. Series of $e^x$, $\sin x$, $\ln(1+x)$ etc. are in 02 · Differential Calculus §8.

$$\sum_{n=0}^\infty x^n = \frac1{1-x}, \quad \sum_{n=1}^\infty nx^{n-1} = \frac{1}{(1-x)^2}, \quad \sum_{n=1}^\infty\frac{x^n}{n} = -\ln(1-x) \qquad (|x|<1)$$

$$\sum_{n=1}^\infty\frac1{n^2} = \frac{\pi^2}6, \qquad \sum_{n=1}^\infty\frac{(-1)^{n+1}}{n} = \ln2, \qquad \sum_{n=0}^\infty\frac{(-1)^n}{2n+1} = \frac\pi4$$


## 4. Fourier series (period $2L$)

$$f(x) = \frac{a_0}2 + \sum_{n=1}^\infty\left(a_n\cos\frac{n\pi x}{L} + b_n\sin\frac{n\pi x}L\right)$$

$$a_0 = \frac1L\int_{-L}^{L}f(x)dx, \qquad a_n = \frac1L\int_{-L}^Lf(x)\cos\frac{n\pi x}Ldx, \qquad b_n = \frac1L\int_{-L}^Lf(x)\sin\frac{n\pi x}Ldx$$

For period $2\pi$ ($L=\pi$): $a_n = \frac1\pi\int_{-\pi}^\pi f(x)\cos nx\,dx$, $b_n = \frac1\pi\int_{-\pi}^\pi f(x)\sin nx\,dx$.

**Even function:** only cosine terms ($b_n = 0$). **Odd function:** only sine terms ($a_0=a_n=0$).

**Half-range series** on $(0,L)$: cosine series uses $a_n = \frac2L\int_0^Lf\cos\frac{n\pi x}L dx$; sine series uses $b_n = \frac2L\int_0^Lf\sin\frac{n\pi x}Ldx$.

**Dirichlet conditions:** at a jump, the series converges to $\frac{f(x^+)+f(x^-)}{2}$. Near a jump the partial sums overshoot by about 9% (Gibbs phenomenon).


## 5. Complex form and Parseval

$$f(x) = \sum_{n=-\infty}^{\infty}c_ne^{in\pi x/L}, \qquad c_n = \frac{1}{2L}\int_{-L}^{L}f(x)e^{-in\pi x/L}dx$$

**Parseval's identity:** $\frac1L\int_{-L}^L f^2\,dx = \frac{a_0^2}{2} + \sum_{n=1}^{\infty}(a_n^2+b_n^2)$

**Orthogonality:** $\int_{-\pi}^{\pi}\sin mx\sin nx\,dx = \int_{-\pi}^\pi\cos mx\cos nx\,dx = \pi\delta_{mn}$ $(m,n\ge1)$, $\int_{-\pi}^\pi\sin mx\cos nx\,dx = 0$


## 6. Harmonic analysis (from tabulated data)

With $N$ equally spaced samples $y_k$ over one period: $a_0 = \frac2N\sum y_k$, $a_n = \frac2N\sum y_k\cos nx_k$, $b_n = \frac2N\sum y_k\sin nx_k$.

Amplitude of the $n$-th harmonic: $\sqrt{a_n^2+b_n^2}$.

**ML note:** sinusoidal position encodings, random Fourier features and Fourier neural operators all build on this.
