# 08 · Complex Numbers and Complex Analysis

- **Level:** Intermediate
- **Typical semester:** Semester 1 (complex numbers) and Semester 4 (complex analysis)
- **Prerequisites:** 01, 02, 03
- **Used later in:** Fourier transforms, signal processing, rotary position embeddings (RoPE), stability of state-space models, complex-valued networks.


## 1. Complex numbers

$z = x + iy = re^{i\theta}$, $\;|z| = r = \sqrt{x^2+y^2}$, $\;\arg z = \theta = \tan^{-1}\frac yx$, $\;\bar z = x - iy$, $\;z\bar z = |z|^2$

**Euler's formula:** $e^{i\theta} = \cos\theta + i\sin\theta$; $\;e^{i\pi} + 1 = 0$

$\cos\theta = \frac{e^{i\theta}+e^{-i\theta}}{2}$, $\;\sin\theta = \frac{e^{i\theta}-e^{-i\theta}}{2i}$

**De Moivre's theorem:** $(\cos\theta + i\sin\theta)^n = \cos n\theta + i\sin n\theta$

**$n$-th roots:** $z^{1/n} = r^{1/n}e^{i(\theta + 2k\pi)/n}$, $k = 0,\dots,n-1$. Roots of unity $\omega_k = e^{2\pi ik/n}$, with $\sum_k\omega_k = 0$.

$|z_1z_2| = |z_1||z_2|$, $\;\arg(z_1z_2) = \arg z_1 + \arg z_2$, $\;|z_1+z_2|\le|z_1|+|z_2|$

**ML note:** RoPE treats pairs of features as complex numbers and multiplies by $e^{im\theta}$, so $\langle q e^{im\theta}, k e^{in\theta}\rangle$ depends only on $m-n$.


## 2. Analytic functions and Cauchy-Riemann equations

$f(z) = u(x,y) + iv(x,y)$ is analytic iff the Cauchy-Riemann equations hold (with continuous partials):
$$u_x = v_y, \qquad u_y = -v_x$$
Polar form: $u_r = \frac1rv_\theta$, $\;v_r = -\frac1ru_\theta$.

Then $f'(z) = u_x + iv_x$, and both $u$ and $v$ are **harmonic**: $\nabla^2u = \nabla^2v = 0$.

**Milne-Thomson method** to build $f$ from $u$: $f'(z) = u_x(z,0) - iu_y(z,0)$, then integrate.


## 3. Elementary complex functions

$e^z = e^x(\cos y + i\sin y)$, $\;\log z = \ln|z| + i\arg z$ (multivalued), $\;z^c = e^{c\log z}$

$\sin z = \sin x\cosh y + i\cos x\sinh y$, $\;\cos z = \cos x\cosh y - i\sin x\sinh y$


## 4. Conformal and bilinear transformations

$w = f(z)$ is conformal (angle-preserving) wherever $f$ is analytic and $f'(z)\ne0$.

Standard maps: $w = z + c$ (translation), $w = az$ (rotation + scaling), $w = 1/z$ (inversion), $w = z^2$, $w = e^z$.

**Bilinear (Möbius) transformation:** $w = \frac{az+b}{cz+d}$, $ad-bc\ne0$; determined by three points via the cross-ratio
$$\frac{(w-w_1)(w_2-w_3)}{(w-w_3)(w_2-w_1)} = \frac{(z-z_1)(z_2-z_3)}{(z-z_3)(z_2-z_1)}$$


## 5. Complex integration

**Cauchy's theorem:** if $f$ is analytic inside and on a closed contour $C$, $\oint_Cf(z)\,dz = 0$.

**Cauchy's integral formula:**
$$f(a) = \frac1{2\pi i}\oint_C\frac{f(z)}{z-a}dz, \qquad f^{(n)}(a) = \frac{n!}{2\pi i}\oint_C\frac{f(z)}{(z-a)^{n+1}}dz$$


## 6. Taylor and Laurent series

Taylor: $f(z) = \sum_{n\ge0}\frac{f^{(n)}(a)}{n!}(z-a)^n$ inside the disc of analyticity.

Laurent: $f(z) = \sum_{n=-\infty}^\infty a_n(z-a)^n$ in an annulus, $a_n = \frac1{2\pi i}\oint\frac{f(z)}{(z-a)^{n+1}}dz$. The negative powers form the principal part.

Singularities: removable, pole of order $m$, essential.


## 7. Residues

**Residue theorem:** $\oint_Cf(z)\,dz = 2\pi i\sum\text{Res}(f, z_k)$ over poles inside $C$.

Simple pole: $\text{Res}(f,a) = \lim_{z\to a}(z-a)f(z)$; for $f = \frac{p}{q}$ with $q(a)=0$: $\text{Res} = \frac{p(a)}{q'(a)}$.

Pole of order $m$: $\text{Res}(f,a) = \frac1{(m-1)!}\lim_{z\to a}\frac{d^{m-1}}{dz^{m-1}}\left[(z-a)^mf(z)\right]$

**Real integrals by contour integration:**
- $\int_0^{2\pi}F(\cos\theta,\sin\theta)\,d\theta$: put $z = e^{i\theta}$, $\cos\theta = \frac{z+z^{-1}}2$, $\sin\theta = \frac{z-z^{-1}}{2i}$, $d\theta = \frac{dz}{iz}$.
- $\int_{-\infty}^\infty\frac{P(x)}{Q(x)}dx = 2\pi i\sum\text{Res in upper half-plane}$ (when $\deg Q\ge\deg P+2$).
- Jordan's lemma for $\int e^{imx}R(x)\,dx$.

**Argument principle / Rouché's theorem:** count zeros inside a contour; used for stability analysis (Nyquist criterion).
