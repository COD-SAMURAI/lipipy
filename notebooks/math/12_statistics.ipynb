# 12 · Statistics, Estimation and Time Series

- **Level:** Intermediate
- **Typical semester:** Semester 3–4 (Probability and Statistics), plus elective statistics courses
- **Prerequisites:** 11
- **Used later in:** MLE is how almost every model is trained; hypothesis tests and confidence intervals are how results are reported; time series for forecasting.


## 1. Descriptive statistics

Mean $\bar x = \frac1n\sum x_i$; median; mode; empirical relation (moderate skew) mean − mode ≈ 3(mean − median)

Sample variance (unbiased) $s^2 = \frac{1}{n-1}\sum(x_i-\bar x)^2$; standard deviation $s$; coefficient of variation $\frac s{\bar x}\times100$

Quartiles, IQR $= Q_3 - Q_1$; outlier rule $x<Q_1 - 1.5\,\text{IQR}$ or $x>Q_3 + 1.5\,\text{IQR}$; z-score $z = \frac{x-\bar x}{s}$

Grouped data: mean $\bar x = A + \frac{\sum f d}{N}h$; median $= L + \frac{N/2 - C}{f}h$; mode $= L + \frac{f_1 - f_0}{2f_1 - f_0 - f_2}h$


## 2. Correlation and regression

**Karl Pearson:** $r = \frac{\sum(x-\bar x)(y - \bar y)}{\sqrt{\sum(x-\bar x)^2\sum(y-\bar y)^2}}$

**Spearman's rank correlation:** $\rho = 1 - \frac{6\sum d_i^2}{n(n^2-1)}$; **Kendall's tau** for ordinal data

**Regression lines:** $y - \bar y = b_{yx}(x - \bar x)$, $\;b_{yx} = r\frac{\sigma_y}{\sigma_x}$; $\;x - \bar x = b_{xy}(y - \bar y)$, $\;b_{xy} = r\frac{\sigma_x}{\sigma_y}$; $\;r = \pm\sqrt{b_{yx}b_{xy}}$; angle between lines $\tan\theta = \frac{1-r^2}{r}\frac{\sigma_x\sigma_y}{\sigma_x^2+\sigma_y^2}$

**Coefficient of determination:** $R^2 = 1 - \frac{SS_{res}}{SS_{tot}}$; adjusted $\bar R^2 = 1 - (1-R^2)\frac{n-1}{n-p-1}$


## 3. Sampling distributions

$\bar X\sim\mathcal N\left(\mu, \frac{\sigma^2}{n}\right)$ (exactly for normal data, approximately by the CLT); standard error $\text{SE} = \frac{\sigma}{\sqrt n}$

$\frac{(n-1)s^2}{\sigma^2}\sim\chi^2_{n-1}$, $\;\frac{\bar X - \mu}{s/\sqrt n}\sim t_{n-1}$, $\;\frac{s_1^2/\sigma_1^2}{s_2^2/\sigma_2^2}\sim F_{n_1-1,n_2-1}$


## 4. Point estimation

**Bias** $\mathbb E[\hat\theta] - \theta$; **MSE** $= \text{Var}(\hat\theta) + \text{Bias}^2$; consistency, efficiency, sufficiency

**Method of moments:** match sample moments to population moments.

**Maximum likelihood (MLE):** $\hat\theta = \arg\max_\theta\sum_i\log p(x_i|\theta)$
- Normal: $\hat\mu = \bar x$, $\hat\sigma^2 = \frac1n\sum(x_i - \bar x)^2$ (biased)
- Bernoulli: $\hat p = \bar x$; Poisson: $\hat\lambda = \bar x$; Exponential: $\hat\lambda = 1/\bar x$
- Invariance: the MLE of $g(\theta)$ is $g(\hat\theta)$

**Fisher information:** $I(\theta) = \mathbb E\left[\left(\frac{\partial}{\partial\theta}\log p(X|\theta)\right)^2\right] = -\mathbb E\left[\frac{\partial^2}{\partial\theta^2}\log p\right]$

**Cramér-Rao bound:** $\text{Var}(\hat\theta)\ge\frac{1}{nI(\theta)}$ for unbiased $\hat\theta$. MLE is asymptotically $\mathcal N(\theta, (nI)^{-1})$.

**ML note:** minimizing cross-entropy loss = maximizing likelihood. Minimizing MSE = MLE under Gaussian noise. The Fisher matrix is the metric used by natural gradient descent.


## 5. Bayesian estimation

$$p(\theta|D) = \frac{p(D|\theta)p(\theta)}{p(D)}, \qquad p(D) = \int p(D|\theta)p(\theta)d\theta \text{ (evidence)}$$

**MAP:** $\hat\theta = \arg\max\left[\log p(D|\theta) + \log p(\theta)\right]$. Gaussian prior gives L2 regularization; Laplace prior gives L1.

**Posterior predictive:** $p(x^*|D) = \int p(x^*|\theta)p(\theta|D)d\theta$

**Conjugate pairs:**

| Likelihood | Prior | Posterior |
|---|---|---|
| Bernoulli / Binomial | Beta$(\alpha,\beta)$ | Beta$(\alpha + k, \beta + n - k)$ |
| Categorical / Multinomial | Dirichlet$(\alpha)$ | Dirichlet$(\alpha + \text{counts})$ |
| Poisson | Gamma$(\alpha,\beta)$ | Gamma$(\alpha + \sum x, \beta + n)$ |
| Normal (known $\sigma^2$) | Normal$(\mu_0,\tau^2)$ | Normal, precision $\frac1{\tau^2} + \frac{n}{\sigma^2}$ |

**Bayesian model comparison:** Bayes factor $\frac{p(D|M_1)}{p(D|M_2)}$; BIC $= k\ln n - 2\ln\hat L$; AIC $= 2k - 2\ln\hat L$


## 6. Interval estimation

Mean, known $\sigma$: $\bar x\pm z_{\alpha/2}\frac{\sigma}{\sqrt n}$ ($z_{0.025} = 1.96$); unknown $\sigma$: $\bar x\pm t_{\alpha/2,n-1}\frac{s}{\sqrt n}$

Proportion: $\hat p\pm z_{\alpha/2}\sqrt{\frac{\hat p(1-\hat p)}n}$; Wilson interval is better for small $n$.

**Bootstrap:** resample the data with replacement $B$ times, recompute the statistic, take percentiles.


## 7. Hypothesis testing

Null $H_0$, alternative $H_1$; Type I error $\alpha = P(\text{reject }H_0|H_0)$; Type II $\beta$; power $1-\beta$; p-value = probability of a result at least this extreme under $H_0$.

| Test | Statistic | Use |
|---|---|---|
| z-test | $z = \frac{\bar x - \mu_0}{\sigma/\sqrt n}$ | mean, known variance or large $n$ |
| One-sample t | $t = \frac{\bar x-\mu_0}{s/\sqrt n}$, df $n-1$ | mean, small sample |
| Two-sample t | $t = \frac{\bar x_1 - \bar x_2}{s_p\sqrt{1/n_1 + 1/n_2}}$ | two means (pooled) |
| Welch t | $t = \frac{\bar x_1 - \bar x_2}{\sqrt{s_1^2/n_1 + s_2^2/n_2}}$ | unequal variances |
| Paired t | $t = \frac{\bar d}{s_d/\sqrt n}$ | before/after, two models on the same data |
| Two-proportion z | $z = \frac{\hat p_1 - \hat p_2}{\sqrt{\hat p(1-\hat p)(1/n_1+1/n_2)}}$ | A/B tests |
| $\chi^2$ goodness of fit | $\chi^2 = \sum\frac{(O-E)^2}{E}$ | distribution fit |
| $\chi^2$ independence | same, df $(r-1)(c-1)$ | contingency tables |
| F-test | $F = s_1^2/s_2^2$ | equal variances |
| ANOVA | $F = \frac{MS_{between}}{MS_{within}}$ | 3+ group means |
| Mann-Whitney U, Wilcoxon signed-rank, Kruskal-Wallis | rank-based | non-parametric |
| McNemar | $\frac{(b-c)^2}{b+c}$ | two classifiers on the same test set |
| Kolmogorov-Smirnov | $\sup|F_1 - F_2|$ | compare distributions |

**Multiple comparisons:** Bonferroni $\alpha/m$; Benjamini-Hochberg controls the false discovery rate.


## 8. Linear models (statistical view)

$y = X\beta + \varepsilon$, $\varepsilon\sim\mathcal N(0,\sigma^2I)$: $\hat\beta = (X^\top X)^{-1}X^\top y$, $\;\text{Var}(\hat\beta) = \sigma^2(X^\top X)^{-1}$

**Gauss-Markov:** OLS is the best linear unbiased estimator (BLUE).

**Bias-variance decomposition:** $\mathbb E[(y - \hat f(x))^2] = \text{Bias}[\hat f]^2 + \text{Var}[\hat f] + \sigma^2$


## 9. Time series

Components: trend, seasonality, cycle, noise. Stationarity: constant mean and autocovariance $\gamma(h)$.

Autocorrelation $\rho(h) = \frac{\gamma(h)}{\gamma(0)}$; partial autocorrelation (PACF).

**Moving average smoothing:** $\hat y_t = \frac1k\sum_{i=0}^{k-1}y_{t-i}$

**Exponential smoothing:** $s_t = \alpha y_t + (1-\alpha)s_{t-1}$. Holt (trend) and Holt-Winters (trend + seasonality).

**AR($p$):** $y_t = c + \sum_{i=1}^p\phi_iy_{t-i} + \varepsilon_t$; **MA($q$):** $y_t = \mu + \varepsilon_t + \sum_{j=1}^q\theta_j\varepsilon_{t-j}$

**ARIMA($p,d,q$):** ARMA on the $d$-times differenced series $\nabla^dy_t$; SARIMA adds seasonal terms.

**GARCH(1,1):** $\sigma_t^2 = \omega + \alpha\varepsilon_{t-1}^2 + \beta\sigma_{t-1}^2$

**Kalman filter** for $x_t = Fx_{t-1} + w$, $y_t = Hx_t + v$, $w\sim\mathcal N(0,Q)$, $v\sim\mathcal N(0,R)$:
- predict: $\hat x^- = F\hat x$, $\;P^- = FPF^\top + Q$
- gain: $K = P^-H^\top(HP^-H^\top + R)^{-1}$
- update: $\hat x = \hat x^- + K(y - H\hat x^-)$, $\;P = (I - KH)P^-$

**Forecast metrics:** MAE, RMSE, MAPE $= \frac{100}{n}\sum\left|\frac{y - \hat y}{y}\right|$, sMAPE, MASE.


## 10. Spatial statistics

**Moran's I:** $I = \frac{n}{W}\frac{\sum_{ij}w_{ij}(x_i - \bar x)(x_j - \bar x)}{\sum_i(x_i-\bar x)^2}$, measuring spatial autocorrelation.

**Variogram:** $\gamma(h) = \frac12\mathbb E[(Z(s+h) - Z(s))^2]$. **Kriging** = best linear unbiased spatial interpolation, the same as Gaussian-process regression.
