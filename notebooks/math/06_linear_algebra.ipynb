# 06 · Linear Algebra

- **Level:** Fundamentals → Intermediate
- **Typical semester:** Semester 1 (matrices, rank, eigenvalues) and Semester 3–4 (vector spaces, linear transformations)
- **Prerequisites:** 01
- **Used later in:** everything in ML. Data is a matrix, a layer is a matrix multiply, PCA is an eigendecomposition, attention is $QK^\top$.


## 1. Vectors

Dot product $u\cdot v = \sum u_iv_i = \|u\|\|v\|\cos\theta$; cosine similarity $\frac{u\cdot v}{\|u\|\|v\|}$

Norms: $\|x\|_1 = \sum|x_i|$, $\|x\|_2 = \sqrt{\sum x_i^2}$, $\|x\|_p = (\sum|x_i|^p)^{1/p}$, $\|x\|_\infty = \max|x_i|$

Projection of $u$ on $v$: $\text{proj}_vu = \frac{u\cdot v}{v\cdot v}v$

Cross product (3D): $u\times v = \begin{vmatrix}\mathbf i & \mathbf j & \mathbf k\\u_1&u_2&u_3\\v_1&v_2&v_3\end{vmatrix}$, $\|u\times v\| = \|u\|\|v\|\sin\theta$; scalar triple product $[u\,v\,w] = u\cdot(v\times w)$ = volume of parallelepiped


## 2. Matrix operations

$(AB)_{ij} = \sum_kA_{ik}B_{kj}$; in general $AB \ne BA$

$(AB)^\top = B^\top A^\top$, $\quad (AB)^{-1} = B^{-1}A^{-1}$, $\quad (A^\top)^{-1} = (A^{-1})^\top$

Trace: $\text{tr}(A) = \sum A_{ii}$, $\;\text{tr}(AB) = \text{tr}(BA)$, $\;\text{tr}(ABC) = \text{tr}(CAB)$

Special matrices: symmetric $A = A^\top$; skew-symmetric $A = -A^\top$; orthogonal $Q^\top Q = I$; Hermitian $A = A^*$; unitary $U^*U = I$; idempotent $A^2=A$; nilpotent $A^k = 0$


## 3. Determinants and inverse

$$\det\begin{pmatrix}a&b\\c&d\end{pmatrix} = ad - bc, \qquad \begin{pmatrix}a&b\\c&d\end{pmatrix}^{-1} = \frac{1}{ad-bc}\begin{pmatrix}d&-b\\-c&a\end{pmatrix}$$

$\det(AB) = \det A\det B$, $\;\det(A^\top) = \det A$, $\;\det(A^{-1}) = 1/\det A$, $\;\det(cA) = c^n\det A$

Cofactor expansion: $\det A = \sum_j a_{ij}C_{ij}$, $C_{ij} = (-1)^{i+j}M_{ij}$; inverse $A^{-1} = \frac{\text{adj}A}{\det A}$

**Cramer's rule:** $x_i = \frac{\det A_i}{\det A}$

**Matrix determinant lemma:** $\det(A + uv^\top) = (1 + v^\top A^{-1}u)\det A$

**Sherman-Morrison:** $(A+uv^\top)^{-1} = A^{-1} - \frac{A^{-1}uv^\top A^{-1}}{1+v^\top A^{-1}u}$

**Woodbury identity:** $(A + UCV)^{-1} = A^{-1} - A^{-1}U(C^{-1}+VA^{-1}U)^{-1}VA^{-1}$


## 4. Rank and systems of linear equations

Rank = number of non-zero rows in row-echelon form = dimension of the column space.

$Ax = b$ with $A$ of size $m\times n$: consistent iff $\text{rank}(A) = \text{rank}([A|b])$; unique solution iff that rank $= n$; infinitely many if $< n$.

**Rank-nullity theorem:** $\text{rank}(A) + \text{nullity}(A) = n$

Methods: Gauss elimination, Gauss-Jordan, LU decomposition (see 10 · Numerical methods).


## 5. Vector spaces

Subspace, span, linear independence ($\sum c_iv_i = 0 \Rightarrow c_i = 0$), basis, dimension.

Four fundamental subspaces of $A$: column space, null space, row space, left null space. $\text{Null}(A) \perp \text{Row}(A)$.

**Gram-Schmidt:** $u_k = v_k - \sum_{j<k}\text{proj}_{u_j}v_k$, then normalize, giving an orthonormal basis.


## 6. Eigenvalues and eigenvectors

$$Av = \lambda v, \qquad \det(A-\lambda I) = 0$$

$\sum\lambda_i = \text{tr}(A)$, $\;\prod\lambda_i = \det A$. Eigenvalues of $A^k$ are $\lambda^k$, of $A^{-1}$ are $1/\lambda$.

**Cayley-Hamilton:** every matrix satisfies its own characteristic equation. Used to find $A^{-1}$ and $A^n$.

Symmetric matrices have real eigenvalues and orthogonal eigenvectors.

**Diagonalization:** $A = PDP^{-1}$; for symmetric $A$, $A = Q\Lambda Q^\top$ (spectral theorem). Then $A^n = PD^nP^{-1}$.

**Power method:** $x_{k+1} = \frac{Ax_k}{\|Ax_k\|}$ converges to the dominant eigenvector (used in PageRank and spectral norm estimation).

**Rayleigh quotient:** $R(x) = \frac{x^\top Ax}{x^\top x}$, with $\lambda_{\min} \le R(x) \le \lambda_{\max}$.


## 7. Quadratic forms and definiteness

$Q(x) = x^\top Ax$ ($A$ symmetric). Positive definite iff all eigenvalues $>0$ iff all leading principal minors $>0$ (Sylvester's criterion).

Reduction to canonical form by orthogonal transformation $x = Py$: $Q = \sum\lambda_iy_i^2$. Index = number of positive terms, signature = positives minus negatives.

**ML note:** a Hessian that is positive definite means a local minimum. Covariance matrices are positive semi-definite.


## 8. Matrix decompositions

| Decomposition | Form | Where it is used |
|---|---|---|
| LU | $A = LU$ (with pivoting $PA = LU$) | solving linear systems |
| Cholesky | $A = LL^\top$ ($A$ positive definite) | Gaussian processes, sampling $\mathcal N(\mu,\Sigma)$ |
| QR | $A = QR$ | least squares, eigenvalue algorithm |
| Eigendecomposition | $A = Q\Lambda Q^\top$ | PCA, spectral clustering |
| SVD | $A = U\Sigma V^\top$ | PCA, low-rank approximation, LoRA, pseudo-inverse |
| Schur | $A = QTQ^*$ | numerical eigenvalues |
| Jordan | $A = PJP^{-1}$ | theory of non-diagonalizable matrices |

**Eckart-Young theorem:** the best rank-$k$ approximation of $A$ (in Frobenius or spectral norm) is $A_k = \sum_{i=1}^k\sigma_iu_iv_i^\top$.

**Moore-Penrose pseudo-inverse:** $A^+ = V\Sigma^+U^\top$; least-squares solution $x = A^+b = (A^\top A)^{-1}A^\top b$ when $A$ has full column rank.


## 9. Matrix norms and conditioning

Frobenius $\|A\|_F = \sqrt{\sum a_{ij}^2} = \sqrt{\text{tr}(A^\top A)} = \sqrt{\sum\sigma_i^2}$

Spectral $\|A\|_2 = \sigma_{\max}(A)$; nuclear $\|A\|_* = \sum\sigma_i$

Condition number $\kappa(A) = \|A\|\|A^{-1}\| = \frac{\sigma_{\max}}{\sigma_{\min}}$. Large $\kappa$ means slow gradient descent and unstable solutions.


## 10. Linear transformations

$T(\alpha u + \beta v) = \alpha T(u) + \beta T(v)$. Rotation in 2D: $R_\theta = \begin{pmatrix}\cos\theta & -\sin\theta\\\sin\theta&\cos\theta\end{pmatrix}$. Change of basis: $[T]_B = P^{-1}[T]P$.


## 11. Kronecker product and vectorization

$(A\otimes B)(C\otimes D) = AC\otimes BD$, $\;\text{vec}(AXB) = (B^\top\otimes A)\text{vec}(X)$

Used in K-FAC and Shampoo optimizers.


## 12. Positive-definite kernels

A function $k(x,y)$ is a valid kernel iff every Gram matrix $K_{ij} = k(x_i,x_j)$ is positive semi-definite (Mercer's theorem). Then $k(x,y) = \langle\phi(x),\phi(y)\rangle$ for some feature map $\phi$.
