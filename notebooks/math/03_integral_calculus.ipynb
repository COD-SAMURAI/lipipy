## 1. Standard indefinite integrals

$$\int x^n\,dx = \frac{x^{n+1}}{n+1} + C\;(n\ne-1), \qquad \int \frac{1}{x}\,dx = \ln|x| + C$$

$$\int e^x dx = e^x + C, \qquad \int a^x dx = \frac{a^x}{\ln a} + C$$

$$\int\sin x\,dx = -\cos x, \quad \int\cos x\,dx = \sin x, \quad \int\sec^2x\,dx = \tan x, \quad \int\csc^2x\,dx = -\cot x$$

$$\int\tan x\,dx = \ln|\sec x|, \quad \int\cot x\,dx = \ln|\sin x|, \quad \int\sec x\,dx = \ln|\sec x+\tan x|, \quad \int\csc x\,dx = \ln|\csc x-\cot x|$$

$$\int\frac{dx}{a^2+x^2} = \frac1a\tan^{-1}\frac xa, \qquad \int\frac{dx}{\sqrt{a^2-x^2}} = \sin^{-1}\frac xa$$

$$\int\frac{dx}{x^2-a^2} = \frac{1}{2a}\ln\left|\frac{x-a}{x+a}\right|, \qquad \int\frac{dx}{a^2-x^2} = \frac{1}{2a}\ln\left|\frac{a+x}{a-x}\right|$$

$$\int\frac{dx}{\sqrt{x^2\pm a^2}} = \ln\left|x+\sqrt{x^2\pm a^2}\right|$$

$$\int\sqrt{a^2-x^2}\,dx = \frac x2\sqrt{a^2-x^2} + \frac{a^2}2\sin^{-1}\frac xa, \qquad \int\sqrt{x^2\pm a^2}\,dx = \frac x2\sqrt{x^2\pm a^2} \pm \frac{a^2}2\ln\left|x+\sqrt{x^2\pm a^2}\right|$$

$$\int e^{ax}\sin bx\,dx = \frac{e^{ax}}{a^2+b^2}(a\sin bx - b\cos bx), \qquad \int e^{ax}\cos bx\,dx = \frac{e^{ax}}{a^2+b^2}(a\cos bx + b\sin bx)$$


## 2. Techniques of integration

**Substitution:** $\int f(g(x))g'(x)\,dx = \int f(u)\,du$ with $u = g(x)$. Special case $\int\frac{f'(x)}{f(x)}dx = \ln|f(x)|$.

**By parts:** $\int u\,dv = uv - \int v\,du$. Choose $u$ by ILATE order: Inverse trig, Log, Algebraic, Trig, Exponential.

**Repeated parts (Bernoulli's formula):** $\int uv\,dx = uv_1 - u'v_2 + u''v_3 - \cdots$, where $v_k$ are successive integrals of $v$.

**Partial fractions:** $\frac{P(x)}{(x-a)(x-b)} = \frac{A}{x-a} + \frac{B}{x-b}$; repeated factor $(x-a)^2$ gives $\frac{A}{x-a} + \frac{B}{(x-a)^2}$; irreducible $x^2+px+q$ gives $\frac{Ax+B}{x^2+px+q}$.

**Trigonometric substitutions:** $\sqrt{a^2-x^2}\to x = a\sin\theta$; $\;\sqrt{a^2+x^2}\to x = a\tan\theta$; $\;\sqrt{x^2-a^2}\to x = a\sec\theta$.

**Weierstrass substitution:** $t = \tan\frac x2$, $\sin x = \frac{2t}{1+t^2}$, $\cos x = \frac{1-t^2}{1+t^2}$, $dx = \frac{2\,dt}{1+t^2}$.


## 3. Definite integrals and properties

**Fundamental theorem of calculus:** $\int_a^b f(x)\,dx = F(b) - F(a)$ where $F' = f$, and $\frac{d}{dx}\int_a^x f(t)\,dt = f(x)$.

$$\int_a^b f(x)dx = \int_a^b f(a+b-x)dx, \qquad \int_0^a f(x)dx = \int_0^a f(a-x)dx$$

$$\int_{-a}^{a}f(x)dx = \begin{cases}2\int_0^a f(x)dx & f\text{ even}\\ 0 & f\text{ odd}\end{cases}$$

**Leibniz rule (differentiation under the integral sign):**
$$\frac{d}{dx}\int_{a(x)}^{b(x)} f(x,t)\,dt = f(x,b)\,b'(x) - f(x,a)\,a'(x) + \int_{a(x)}^{b(x)}\frac{\partial f}{\partial x}dt$$

**ML note:** the "reparameterization trick" in VAEs relies on moving a derivative inside an expectation (an integral).


## 4. Reduction formulas (Wallis)

$$\int_0^{\pi/2}\sin^n x\,dx = \int_0^{\pi/2}\cos^nx\,dx = \begin{cases}\frac{(n-1)(n-3)\cdots 1}{n(n-2)\cdots 2}\cdot\frac\pi2 & n \text{ even}\\[4pt] \frac{(n-1)(n-3)\cdots 2}{n(n-2)\cdots 3} & n\text{ odd}\end{cases}$$

$$\int_0^{\pi/2}\sin^mx\cos^nx\,dx = \frac{[(m-1)(m-3)\cdots][(n-1)(n-3)\cdots]}{(m+n)(m+n-2)\cdots}\cdot k, \quad k = \frac\pi2 \text{ if } m,n \text{ both even, else } 1$$


## 5. Improper integrals and special functions

**Gamma function:** $\Gamma(n) = \int_0^\infty x^{n-1}e^{-x}dx$, $\quad \Gamma(n+1) = n\Gamma(n)$, $\quad \Gamma(n+1) = n!$, $\quad \Gamma(\tfrac12) = \sqrt\pi$

**Beta function:** $B(m,n) = \int_0^1 x^{m-1}(1-x)^{n-1}dx = \frac{\Gamma(m)\Gamma(n)}{\Gamma(m+n)} = 2\int_0^{\pi/2}\sin^{2m-1}\theta\cos^{2n-1}\theta\,d\theta$

**Gaussian integral:** $\int_{-\infty}^{\infty}e^{-x^2}dx = \sqrt\pi$, $\quad \int_{-\infty}^\infty e^{-ax^2}dx = \sqrt{\frac\pi a}$, $\quad \int_{-\infty}^\infty e^{-\frac{(x-\mu)^2}{2\sigma^2}}dx = \sigma\sqrt{2\pi}$

**Error function:** $\text{erf}(x) = \frac{2}{\sqrt\pi}\int_0^xe^{-t^2}dt$ (used in GELU: $\text{GELU}(x) = x\,\Phi(x) = \frac x2\left[1+\text{erf}\left(\frac{x}{\sqrt2}\right)\right]$)

**Stirling's approximation:** $n! \approx \sqrt{2\pi n}\left(\frac ne\right)^n$, $\quad \ln n! \approx n\ln n - n$


## 6. Applications of single integrals

**Area between curves:** $A = \int_a^b |f(x) - g(x)|\,dx$; polar area $A = \frac12\int_\alpha^\beta r^2\,d\theta$

**Arc length:** $L = \int_a^b\sqrt{1+\left(\frac{dy}{dx}\right)^2}dx$; parametric $L = \int\sqrt{\dot x^2+\dot y^2}\,dt$; polar $L = \int\sqrt{r^2+\left(\frac{dr}{d\theta}\right)^2}d\theta$

**Volume of revolution (about $x$-axis):** disc $V = \pi\int_a^b y^2\,dx$; shell (about $y$-axis) $V = 2\pi\int_a^b x\,y\,dx$

**Surface of revolution:** $S = 2\pi\int_a^b y\sqrt{1+y'^2}\,dx$

**Mean value of a function:** $\bar f = \frac{1}{b-a}\int_a^b f(x)\,dx$; RMS value $\sqrt{\frac{1}{b-a}\int_a^b f^2\,dx}$


## 7. Multiple integrals

**Double integral:** $\iint_R f(x,y)\,dA = \int_a^b\int_{g_1(x)}^{g_2(x)}f(x,y)\,dy\,dx$ (Fubini: order can be changed for continuous $f$; change the limits accordingly)

**Change of variables:** $\iint f(x,y)\,dx\,dy = \iint f(x(u,v),y(u,v))\left|\frac{\partial(x,y)}{\partial(u,v)}\right|du\,dv$

Polar: $dx\,dy = r\,dr\,d\theta$; cylindrical: $dV = r\,dr\,d\theta\,dz$; spherical: $dV = r^2\sin\theta\,dr\,d\theta\,d\phi$

**Triple integral:** $\iiint_V f\,dV$; volume $V = \iiint_V dV$

**Area and mass:** $A = \iint_R dA$, mass $M = \iint\rho\,dA$, centroid $\bar x = \frac1M\iint x\rho\,dA$, moment of inertia $I_x = \iint y^2\rho\,dA$

**ML note:** the change-of-variables formula (with $|\det J|$) is the core of normalizing flows.
