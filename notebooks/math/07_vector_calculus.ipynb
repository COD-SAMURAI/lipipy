# 07 · Vector Calculus

- **Level:** Intermediate
- **Typical semester:** Semester 2 (Engineering Mathematics II)
- **Prerequisites:** 02, 03, 06
- **Used later in:** gradients of multivariate losses, divergence in normalizing flows and score matching, physics-informed ML.


## 1. Vector differentiation

For $\mathbf r(t) = x\mathbf i + y\mathbf j + z\mathbf k$: velocity $\mathbf v = \frac{d\mathbf r}{dt}$, acceleration $\mathbf a = \frac{d^2\mathbf r}{dt^2}$.

$\frac{d}{dt}(\mathbf u\cdot\mathbf v) = \mathbf u'\cdot\mathbf v + \mathbf u\cdot\mathbf v'$, $\quad\frac{d}{dt}(\mathbf u\times\mathbf v) = \mathbf u'\times\mathbf v + \mathbf u\times\mathbf v'$

Unit tangent $\mathbf T = \frac{\mathbf r'}{|\mathbf r'|}$; curvature $\kappa = \frac{|\mathbf r'\times\mathbf r''|}{|\mathbf r'|^3}$; torsion $\tau = \frac{[\mathbf r'\,\mathbf r''\,\mathbf r''']}{|\mathbf r'\times\mathbf r''|^2}$


## 2. Del operator: gradient, divergence, curl, Laplacian

$$\nabla = \mathbf i\frac{\partial}{\partial x} + \mathbf j\frac\partial{\partial y} + \mathbf k\frac{\partial}{\partial z}$$

**Gradient** of scalar $\phi$: $\nabla\phi = \phi_x\mathbf i + \phi_y\mathbf j + \phi_z\mathbf k$, pointing in the direction of steepest increase.

**Directional derivative** along unit $\hat a$: $D_{\hat a}\phi = \nabla\phi\cdot\hat a$; maximum value $|\nabla\phi|$.

**Unit normal to surface** $\phi = c$: $\hat n = \frac{\nabla\phi}{|\nabla\phi|}$; angle between surfaces: $\cos\theta = \frac{\nabla\phi_1\cdot\nabla\phi_2}{|\nabla\phi_1||\nabla\phi_2|}$

**Divergence** of $\mathbf F = F_1\mathbf i + F_2\mathbf j + F_3\mathbf k$: $\nabla\cdot\mathbf F = \frac{\partial F_1}{\partial x} + \frac{\partial F_2}{\partial y} + \frac{\partial F_3}{\partial z}$. Solenoidal if $\nabla\cdot\mathbf F = 0$.

**Curl:** $\nabla\times\mathbf F = \begin{vmatrix}\mathbf i&\mathbf j&\mathbf k\\\partial_x&\partial_y&\partial_z\\F_1&F_2&F_3\end{vmatrix}$. Irrotational if $\nabla\times\mathbf F = 0$, then $\mathbf F = \nabla\phi$ (conservative, $\phi$ = scalar potential).

**Laplacian:** $\nabla^2\phi = \phi_{xx} + \phi_{yy} + \phi_{zz}$


## 3. Vector identities

$$\nabla\times(\nabla\phi) = 0, \qquad \nabla\cdot(\nabla\times\mathbf F) = 0$$
$$\nabla\cdot(\phi\mathbf F) = \nabla\phi\cdot\mathbf F + \phi\nabla\cdot\mathbf F, \qquad \nabla\times(\phi\mathbf F) = \nabla\phi\times\mathbf F + \phi\nabla\times\mathbf F$$
$$\nabla\times(\nabla\times\mathbf F) = \nabla(\nabla\cdot\mathbf F) - \nabla^2\mathbf F$$
$$\nabla\cdot(\mathbf F\times\mathbf G) = \mathbf G\cdot(\nabla\times\mathbf F) - \mathbf F\cdot(\nabla\times\mathbf G)$$
$$\nabla r^n = nr^{n-2}\mathbf r, \qquad \nabla\cdot\mathbf r = 3, \qquad \nabla\times\mathbf r = 0, \qquad \nabla^2 r^n = n(n+1)r^{n-2}$$


## 4. Line, surface and volume integrals

**Line integral (work):** $\int_C\mathbf F\cdot d\mathbf r = \int_C(F_1dx + F_2dy + F_3dz)$. Path-independent iff $\mathbf F = \nabla\phi$, then $= \phi(B) - \phi(A)$.

**Surface integral (flux):** $\iint_S\mathbf F\cdot\hat n\,dS$. For projection on the $xy$-plane: $dS = \frac{dx\,dy}{|\hat n\cdot\mathbf k|}$.

**Volume integral:** $\iiint_V\phi\,dV$.


## 5. Integral theorems

**Green's theorem (plane):** $\oint_C(M\,dx + N\,dy) = \iint_R\left(\frac{\partial N}{\partial x} - \frac{\partial M}{\partial y}\right)dx\,dy$

Area enclosed: $A = \frac12\oint_C(x\,dy - y\,dx)$

**Stokes' theorem:** $\oint_C\mathbf F\cdot d\mathbf r = \iint_S(\nabla\times\mathbf F)\cdot\hat n\,dS$

**Gauss divergence theorem:** $\iint_S\mathbf F\cdot\hat n\,dS = \iiint_V\nabla\cdot\mathbf F\,dV$


## 6. Curvilinear coordinates

Scale factors: cylindrical $(h_\rho,h_\phi,h_z) = (1,\rho,1)$; spherical $(h_r,h_\theta,h_\phi) = (1,r,r\sin\theta)$.

$$\nabla^2\phi = \frac{1}{h_1h_2h_3}\left[\frac{\partial}{\partial u_1}\left(\frac{h_2h_3}{h_1}\frac{\partial\phi}{\partial u_1}\right) + \frac{\partial}{\partial u_2}\left(\frac{h_3h_1}{h_2}\frac{\partial\phi}{\partial u_2}\right)+\frac{\partial}{\partial u_3}\left(\frac{h_1h_2}{h_3}\frac{\partial\phi}{\partial u_3}\right)\right]$$


## 7. Bridge to ML

- **Gradient descent** moves along $-\nabla L$, the direction of steepest decrease.
- **Continuity equation** $\partial_tp + \nabla\cdot(p\,\mathbf v) = 0$ describes how a probability density moves under a velocity field: the basis of continuous normalizing flows and flow matching.
- **Instantaneous change of variables:** $\frac{d\log p(\mathbf z_t)}{dt} = -\nabla\cdot\mathbf f(\mathbf z_t)$ (FFJORD / neural ODE flows).
- **Score matching** uses $\text{tr}(\nabla_x s_\theta(x)) = \nabla\cdot s_\theta$; Hutchinson's estimator $\text{tr}(A) = \mathbb E[\epsilon^\top A\epsilon]$ makes it cheap.
