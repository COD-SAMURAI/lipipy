## 1. PAC learning

A class $\mathcal H$ is PAC-learnable if for all $\varepsilon,\delta$ there is $m(\varepsilon,\delta)$ such that with probability $\ge1 - \delta$ the learned $h$ has $R(h)\le\varepsilon$ (realizable) or $R(h)\le\min_{\mathcal H}R + \varepsilon$ (agnostic).

Finite class, realizable: $m\ge\frac1\varepsilon\left(\ln|\mathcal H| + \ln\frac1\delta\right)$. Agnostic: $m\ge\frac{1}{2\varepsilon^2}\left(\ln|\mathcal H| + \ln\frac2\delta\right)$.


## 2. VC dimension

$d_{VC}(\mathcal H)$ = size of the largest set $\mathcal H$ can shatter. Linear classifiers in $\mathbb R^d$: $d + 1$.

**Sauer-Shelah lemma:** growth function $\Pi_\mathcal H(m)\le\sum_{i=0}^{d}\binom mi\le\left(\frac{em}{d}\right)^d$

**VC generalization bound:** with probability $1 - \delta$,
$$R(h)\le\hat R(h) + \sqrt{\frac{8}{m}\left(d\ln\frac{2em}{d} + \ln\frac4\delta\right)}$$


## 3. Rademacher complexity

$\hat{\mathfrak R}_S(\mathcal F) = \mathbb E_\sigma\left[\sup_{f\in\mathcal F}\frac1m\sum_i\sigma_if(x_i)\right]$, $\sigma_i\in\{\pm1\}$ uniform.

Bound: $R(f)\le\hat R(f) + 2\mathfrak R_m(\mathcal F) + \sqrt{\frac{\ln(1/\delta)}{2m}}$. Linear functions with $\|w\|\le B$, $\|x\|\le X$: $\mathfrak R\le\frac{BX}{\sqrt m}$.

**Margin bounds** for neural networks scale with products of layer norms (spectrally normalized margin bounds).


## 4. PAC-Bayes

For any prior $P$ chosen before seeing data and any posterior $Q$, with probability $1 - \delta$:
$$\mathbb E_{h\sim Q}R(h)\le\mathbb E_{h\sim Q}\hat R(h) + \sqrt{\frac{KL(Q\|P) + \ln\frac{2\sqrt m}{\delta}}{2m}}$$
Gives non-vacuous bounds for some trained neural networks; connects flat minima to generalization.


## 5. No free lunch, bias-complexity and Occam

**No free lunch:** averaged over all target functions, every learner has the same off-training-set error. Learning requires inductive bias.

**Occam's razor bound:** with a prefix code over hypotheses of length $|h|$ bits, $R(h)\le\hat R(h) + \sqrt{\frac{|h|\ln2 + \ln(1/\delta)}{2m}}$.


## 6. Deep learning theory

**Neural tangent kernel:** $\Theta(x,x') = \nabla_\theta f(x)^\top\nabla_\theta f(x')$. In the infinite-width limit with NTK parameterization, $\Theta$ stays constant during training and $f_t = f_0 + \Theta(\cdot, X)\Theta^{-1}(I - e^{-\eta\Theta t})(y - f_0(X))$ (kernel regression).

**Mean-field / $\mu$P (maximal update parameterization):** scale initializations and learning rates with width so features keep learning as width grows; hyperparameters transfer from small to large models.

**Double descent:** test error peaks at the interpolation threshold (parameters ≈ samples) and falls again beyond it. **Benign overfitting:** interpolating noisy data can still generalize when the covariance spectrum has many small directions.

**Implicit regularization:** gradient descent on separable logistic regression converges in direction to the max-margin solution $\frac{w_t}{\|w_t\|}\to\frac{w_{SVM}}{\|w_{SVM}\|}$; on underdetermined least squares from zero init it finds the minimum-norm solution.

**Edge of stability:** with step size $\eta$, full-batch GD drives the top Hessian eigenvalue to about $\frac{2}{\eta}$ and hovers there.

**Grokking:** generalization appears long after training accuracy reaches 100%, often linked to weight-norm decay.

**Information-theoretic generalization:** $|\mathbb E[R - \hat R]|\le\sqrt{\frac{2\sigma^2}{m}I(W;S)}$ ($\sigma$-sub-Gaussian loss).


## 7. Scaling and emergence (empirical laws)

Power-law loss in parameters, data and compute (see 19 §11). Data-constrained scaling: repeated epochs lose value roughly after about 4 repetitions. Many "emergent" abilities look sharp under discontinuous metrics (exact match) and smooth under continuous ones (log-likelihood).


## 8. Algorithmic information theory

**Kolmogorov complexity:** $K_U(x) = \min\{|p| : U(p) = x\}$; invariance theorem $|K_U(x) - K_V(x)|\le c_{UV}$. Uncomputable.

**Solomonoff's universal prior:** $M(x) = \sum_{p:\,U(p) = x*}2^{-|p|}$. Prediction $M(x_{t+1}|x_{1:t})$ converges rapidly to any computable environment: total expected squared error $\le\frac{\ln2}{2}K(\mu)$.

**Minimum description length** (see 14 §7) is the practical, computable version.


## 9. Universal intelligence and AIXI

**Legg-Hutter universal intelligence:** $\Upsilon(\pi) = \sum_{\mu\in E}2^{-K(\mu)}V_\mu^\pi$, the expected performance of policy $\pi$ over all computable environments weighted by simplicity.

**AIXI:** the agent that acts optimally under the Solomonoff mixture:
$$a_t = \arg\max_{a_t}\sum_{o_tr_t}\cdots\max_{a_m}\sum_{o_mr_m}[r_t + \cdots + r_m]\sum_{q:\,U(q, a_{1:m}) = o_1r_1\dots o_mr_m}2^{-|q|}$$
Incomputable; approximations include MC-AIXI-CTW (Monte Carlo tree search + context-tree weighting).


## 10. Bayesian brain, predictive coding and active inference

**Variational free energy:** $F = \mathbb E_{q(s)}[\log q(s) - \log p(o, s)] = D_{KL}(q(s)\|p(s|o)) - \log p(o)\ge-\log p(o)$ (the negative ELBO, an upper bound on surprise)

**Predictive coding:** each layer predicts the one below; prediction errors $\varepsilon_l = x_l - f(W_lx_{l+1})$ drive updates of states and weights.

**Active inference:** choose policies minimizing expected free energy
$$G(\pi) = \underbrace{-\mathbb E_{q}[\ln p(o|C)]}_{\text{pragmatic (preferences)}}\;\underbrace{-\;\mathbb E_q\left[D_{KL}(q(s|o,\pi)\|q(s|\pi))\right]}_{\text{epistemic (information gain)}}$$


## 11. Causality

**Structural causal model:** $X_i := f_i(\text{PA}_i, U_i)$. Interventions $do(X = x)$ replace the equation for $X$.

**Back-door adjustment:** if $Z$ blocks all back-door paths, $P(y|do(x)) = \sum_zP(y|x,z)P(z)$. **Front-door:** $P(y|do(x)) = \sum_mP(m|x)\sum_{x'}P(y|m,x')P(x')$.

**Do-calculus** (Pearl's three rules) decides identifiability. **Counterfactuals:** abduction → action → prediction.

**Treatment effects:** ATE $= \mathbb E[Y(1) - Y(0)]$; inverse propensity weighting $\hat\tau = \frac1n\sum\left(\frac{T_iY_i}{e(X_i)} - \frac{(1 - T_i)Y_i}{1 - e(X_i)}\right)$; doubly robust estimators combine outcome models and propensities.


## 12. Meta-learning and continual learning

**MAML:** $\theta' = \theta - \alpha\nabla_\theta\mathcal L_{\mathcal T}(\theta)$, meta-update $\theta\leftarrow\theta - \beta\nabla_\theta\sum_{\mathcal T}\mathcal L_\mathcal T(\theta')$ (second-order through the inner step). **Reptile:** $\theta\leftarrow\theta + \epsilon(\tilde\theta - \theta)$ after $k$ inner SGD steps.

**Prototypical networks:** $p(y = k|x) = \text{softmax}(-d(f(x), c_k))$, $c_k$ = mean embedding of the support set of class $k$.

**Elastic weight consolidation:** $\mathcal L = \mathcal L_B(\theta) + \sum_i\frac\lambda2F_i(\theta_i - \theta_{A,i}^*)^2$ to avoid catastrophic forgetting.


## 13. Open-endedness and quality-diversity

**Novelty search:** $\rho(x) = \frac1k\sum_{i=1}^k\text{dist}(b(x), b(\mu_i))$ over the $k$ nearest behaviours in an archive.

**MAP-Elites:** keep the best solution per cell of a behaviour-descriptor grid; QD-score $= \sum_{\text{cells}}f(\text{elite})$.

**Learning progress curricula:** sample tasks with probability proportional to $|\Delta\text{performance}|$.


## 14. Measuring general intelligence and capability

**Item response theory:** $P(\text{correct}) = \sigma(a_j(\theta_i - b_j))$, ability $\theta_i$, difficulty $b_j$, discrimination $a_j$; used to build better AI benchmarks.

**Chollet's measure of intelligence:** skill-acquisition efficiency over a scope of tasks, accounting for priors, experience and generalization difficulty (motivates ARC-style benchmarks).

**Elo / Bradley-Terry** for arena-style model rankings (see 22 §11).


## 15. Alignment theory

**Goodhart's law** and reward over-optimization (22 §7).

**Debate:** under idealized assumptions, optimal play in a two-player zero-sum debate judged by a polynomial-time verifier can answer any question in PSPACE.

**Corrigibility and utility indifference:** design $U' = U + \text{correction term}$ so the agent is indifferent to being shut down.

**Weak-to-strong generalization:** performance gap recovered $\text{PGR} = \frac{\text{weak-to-strong} - \text{weak}}{\text{strong ceiling} - \text{weak}}$.


## 16. Open problems (no settled formulas yet)

Sample-efficient learning from few examples; robust out-of-distribution generalization; long-horizon planning and credit assignment; grounding and world models; reliable reasoning and self-verification; continual learning without forgetting; interpretability at scale; scalable oversight. These are the active frontier this repository's `research/`, `agents/` and `interpret/` packages aim to make easier to explore.
