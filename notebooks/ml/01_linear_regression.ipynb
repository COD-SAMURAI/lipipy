# Linear Regression from scratch

This notebook derives ordinary least squares, implements it two ways (closed form and gradient descent), and checks the result against scikit-learn.

## 1. The model

$$\hat y = Xw + b$$

We minimise the mean squared error

$$L(w,b) = \frac{1}{n}\lVert y - Xw - b\rVert_2^2.$$

```python
import numpy as np

from lipipy.ml import LinearRegression

rng = np.random.default_rng(0)
X = rng.normal(size=(200, 3))
true_w = np.array([1.5, -2.0, 0.5])
y = X @ true_w + 4.0 + 0.1 * rng.normal(size=200)
```

## 2. Closed form

Centre $X$ and $y$ to remove the intercept, then set the gradient to zero:

$$X_c^\top X_c\, w = X_c^\top y_c \quad\Rightarrow\quad w = (X_c^\top X_c)^{-1} X_c^\top y_c,\qquad b = \bar y - \bar x^\top w.$$

In code we use an SVD least-squares solver instead of inverting $X^\top X$, which is more stable and also works when columns are collinear.

```python
closed = LinearRegression().fit(X, y)
print("w =", closed.coef_, " b =", closed.intercept_)
```

## 3. Gradient descent

$$\nabla_w L = -\tfrac{2}{n}X^\top(y-\hat y),\qquad \nabla_b L = -\tfrac{2}{n}\textstyle\sum_i (y_i-\hat y_i)$$

Repeat $w \leftarrow w - \eta\,\nabla_w L$ until the loss stops improving.

```python
gd = LinearRegression(solver="gd").fit(X, y)
print("w =", gd.coef_, " b =", gd.intercept_, " iterations =", gd.n_iter_)
```

```python
import matplotlib.pyplot as plt

plt.semilogy(gd.loss_history_)
plt.xlabel("iteration")
plt.ylabel("MSE")
plt.title("Gradient descent loss")
plt.show()
```

## 4. Check against scikit-learn

```python
from sklearn.linear_model import LinearRegression as SkLinearRegression

ref = SkLinearRegression().fit(X, y)
print("max |w diff| closed:", np.abs(closed.coef_ - ref.coef_).max())
print("max |w diff| gd:    ", np.abs(gd.coef_ - ref.coef_).max())
```
